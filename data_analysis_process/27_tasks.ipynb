### Problem 3:

Go to this url: https://www.flipkart.com/search?q=smartphones. This is the url to find phones in flipkart website. You have to extract the below things:
1. image url of the phone
2. name of the image
3. average ratings
4. total ratings
5. total reviews
6. discounted price
7. actual price

Extract all the phones which are available in this website. So you have to use the pagination concept. **Also after requesting every page through the url, please wait for a while (minimum 2-3 seconds), otherwise your IP address can be banned to access the flipkart website later.**

After collecting all the data, save that in a JSON file.

```python
# code here

import requests
from bs4 import BeautifulSoup
import numpy as np
import pandas as pd
import time
```

```python
img_url = []
name = []
avg = []
total = []
rew = []
dis = []
act = []


for j in range(1, 5):
    time.sleep(3)
    url = f'https://www.flipkart.com/search?q=smartphones&page={j}'

    headers = {
    'User-Agent':
    'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    response = requests.get(url, headers=headers)
    print(j, response.status_code)
    soup = BeautifulSoup(response.text, 'html.parser')
    box = soup.find_all('div', class_='nZIRY7')


    for i in box:
        try:
            img_url.append("flipkart.com" + i.find('a', class_='k7wcnx').get('href'))
        except Exception as e:
            print("img", e)
            img_url.append(np.nan)
            
        try:
            name.append(i.find('div', class_='RG5Slk').text)
        except Exception as e:
            print("name", e)
            name.append(np.nan)
            
        try:
            avg.append(i.find('div', class_='MKiFS6').text)
        except Exception as e:
            print("avg", e)
            avg.append(np.nan)
            
        try:
            total.append(i.find('span', class_='PvbNMB').find('span').find('span').text.replace('Ratings', "").strip())
        except Exception as e:
            print("total", e)
            total.append(np.nan)
        
        try:
            review = i.select_one(".PvbNMB > span > span:nth-child(3)")
            rew.append(review.get_text(strip=True).replace("Reviews", "").strip())
        except Exception as e:
            print("rew", e)
            rew.append(np.nan)
        
        try:
            dis.append(i.find('div', class_='hZ3P6w DeU9vF').text)
        except Exception as e:
            print('dis', e)
            dis.append(np.nan)
        
        try:
            act.append(i.find('div', class_='kRYCnD gxR4EY').text)
        except Exception as e:
            print("act", e)
            act.append(np.nan)


```

```python
dic = {
    "img_url":img_url,
    "name":name,
    "avg_rating":avg,
    "total_rating":total,
    "review":rew,
    "discount_price":dis,
    "actual_prices":act
}
```
