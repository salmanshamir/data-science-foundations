# **Cleaning Dataset**

We have to clean the dataset in this order

1. **completeness**

2. **messy data**

3. **validity**

4. **accuracy**

5. **consistency**

# **Define solution**

- we have to remove the writtenby text from author column

- we have to remove the narratedby text from the narrator column

- we have to make columns duration_hour and duration_min

- we have to split the stars columns and make stars and rating columns

- we have to change the encoding of name column

- we have to change the datatype of price to float

- we have to change the datatype of releasedate to datetime

```python
df.info()
```

```python
# code
df["author"] = df['author'].str.split(':').str.get(1).str.strip()
```

```python
# test
df['author']
```

<br>

```python
# code
df["narrator"] = df["narrator"].str.split(":").str.get(1).str.strip()
```

```python
# test
df["narrator"]
```

```python
# code
def hr(text):
    t = text.split('and')
    if len(t)==2:
        hr = t[0].strip().split(' ')[0]
        min = t[1].strip().split(' ')[0]
        return (int(hr)*60)+int(min)
    elif (len(t)==1) & ("Less" in t[0]):
        return int(1)
    elif (len(t)==1) & ("mins" in t[0]):
        min = t[0].strip().split(' ')[0]
        return int(min)
    elif (len(t)==1) & ("hrs" in t[0]):
        hr = t[0].strip().split(' ')[0]
        return int(hr)*60
    else:
        return 'error'
```

```python
df['duration_min'] = df['time'].apply(hr)
```

```python
# test
df.sample(6)
```

```python
# code
def starss(text):
    t = text.split('out')
    return t[0].strip()

df["star"] = df['stars'].apply(starss)


def rat(text):
    t = text.split('stars')
    if t[0]=="Not rated yet":
        return "Not rated yet"
    elif len(t)==2:
        r = t[1].strip().split(' ')[0]
        return r
    else:
        return 'error'
    
df["rating"] = df["stars"].apply(rat)    
```

```python
# test
df.sample(5)
```

```python
# code
def enc(text):
    try:
        return text.encode('latin1').decode('uft-8')
    except:
        return text
    
df["name"] = df['name'].apply(enc)
    
```

```python
# test
df.sample(5)
```
