### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China.

```python
df = pd.read_csv('data/23_data/List of most-polluted cities by particulate matter concentration - List of most-polluted cities by particulate matter concentration.csv')
p = pd.pivot_table(data=df, index="Year", columns="Country", values="PM2.5", aggfunc='sum', fill_value=0)[["China", "Iran"]].reset_index()
plt.plot(p["Year"], p["China"], label="China")
plt.plot(p["Year"], p["Iran"], label="Iran")
plt.xticks(p["Year"])
plt.legend()
plt.title("Year wise PM2.5")
plt.xlabel("Year")
plt.ylabel("PM2.5")
plt.show()
```

### **`Problem-2:`** Draw a histogram of the  column "PM10" of which the y-axis represents the probability (see the documentation how to draw the probability).

```python
# code here

plt.hist(df['PM10'],density=True,bins=50,facecolor='green',edgecolor="black", alpha=0.6)
plt.xlabel('Bins')
plt.ylabel('Probability')
plt.title('Histogram of PM10')
plt.grid()
plt.show()
```

### **`Problem-3:`** Draw a scatter plot where x-axis represents "PM2.5" and y-axis represents "PM10" for two countries Poland and Chile.

```python
# code here
chile_df = df.query("Country == 'Chile'")
poland_df = df.query("Country == 'Poland'")
```

```python
plt.scatter(chile_df['PM2.5'],chile_df['PM10'],marker="+",color='red',label='Chile')
plt.scatter(poland_df['PM2.5'],poland_df['PM10'],marker="D",color='black',label='Poland')
plt.xlabel('PM2.5')
plt.ylabel('PM10')
plt.title('PM2.5 Vs PM10 for Chile and Poland')
plt.legend()
plt.grid()
plt.show()
```

### `Problem-6`

```python
# code here
df = pd.read_csv('data/23_data/company_sales_data - company_sales_data.csv')
df
```

### `Problem-7`
Show sales of each product in march month as pie chart.
* Show Percentage value
* Give Title "Sales in March"
* Explode ToothPaste with shadow

```python
ds = df.melt(id_vars=["month_number", "total_units", "total_profit"], var_name="product", value_name="sales")
d = ds[ds["month_number"]==3]
plt.pie(d["sales"], labels=d["product"], autopct="%0.1f%%", explode=[0, 0, 0.1, 0, 0, 0], shadow=True)
plt.title("Sales in March")
plt.show()
```

```python
# method_2
labels = df[df['month_number'] == 3].iloc[:,1:7].stack().index.get_level_values(1)
values = df[df['month_number'] == 3].iloc[:,1:7].stack().values
```
