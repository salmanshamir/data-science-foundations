# **Annotation**
An annotation is simply text or symbols added to a chart to explain something important.


**Syntax:**

plt.text(x, y, "Your Text")

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
a = [1, 2, 3, 4, 5]
b =  [10, 20, 30, 40, 50]

plt.scatter(a, b)
plt.text(5, 50, "Highest Point", color="red", fontdict={'fontweight':'bold'}, bbox={'facecolor':'yellow', 'boxstyle':'round'})
```

```python
batter = pd.read_csv('data/24_data/batter (1).csv')
batsman = pd.read_csv('data/24_data/batsman_season_record (1).csv')
a = batter.sample(25, random_state=42)


plt.figure(figsize=(20, 10))
plt.scatter(a["runs"], a["avg"], s=500)
for i in range(a.shape[0]):
    plt.text(a["runs"].values[i], a['avg'].values[i], a["batter"].values[i], va='top')
```

# **Subplots**

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 150, 130, 170]
profit = [20, 25, 40, 30, 50]
quantity = [10, 15, 20, 18, 25]

fig, ax = plt.subplots(
    2,
    2,
    figsize=(10, 6)
    )

# Plot 1
ax[0,0].plot(months, sales, marker="o")
ax[0,0].set_title("Sales")

# Plot 2
ax[0,1].bar(months, profit)
ax[0,1].set_title("Profit")

# Plot 3
ax[1,0].plot(months, quantity, marker="o")
ax[1,0].set_title("Quantity")

# Plot 4
ax[1,1].hist(sales)
ax[1,1].set_title("Sales Distribution")

fig.suptitle("Company Performance")

plt.tight_layout()

plt.show()
```

```python
# adjusting space between chart

fig, ax = plt.subplots(
    2,
    2,
    figsize=(10, 6)
    )
# wspace = side wise (left right) space between charts
# vspace = above and below space between charts
plt.subplots_adjust(wspace=0.2, hspace=0.5)

# Plot 1
ax[0,0].plot(months, sales, marker="o")
ax[0,0].set_title("Sales")
ax[0, 0].set_xlabel('month')

# Plot 2
ax[0,1].bar(months, profit)
ax[0,1].set_title("Profit")
ax[0, 1].set_ylabel('profit')

# Plot 3
ax[1,0].plot(months, quantity, marker="o")
ax[1,0].set_title("Quantity")

# Plot 4
ax[1,1].hist(sales)
ax[1,1].set_title("Sales Distribution")

plt.show()
```

# **Line Plot**

```python
sale = pd.read_csv('data/24_data/company_sales_data - company_sales_data.csv')
sale[["month_number", "total_units", "total_profit"]].plot(kind='line', x='month_number')
plt.xticks(ticks=[i for i in range(1, 13)], labels=["jan", "feb", "mar", "april", "may", "june", "july", "aug", "sep", "oct", "nov", 'dec'])
plt.show()
```

```python
# Subplots
sale[["month_number", "total_units", "total_profit"]].plot(kind='line', x='month_number', subplots=True)
plt.xticks(ticks=[i for i in range(1, 13)], labels=["jan", "feb", "mar", "april", "may", "june", "july", "aug", "sep", "oct", "nov", 'dec'])
plt.show()
```
