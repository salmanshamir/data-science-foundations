# **Array Vs List**

### **List**

- A list is a built-in Python data structure used to store a collection of objects.

- A list can contain different types of data.

- A Python list stores references (addresses) to separate Python objects.
- Slow Speed

### **Array**

- A NumPy array is a data structure designed to store large amounts of numerical data efficiently.

- Unlike lists, a NumPy array stores only one data type.

- A NumPy array stores values in one continuous block of memory.
  
- High Speed

```python
a = [i for i in range(10000000)]
b = [i for i in range(10000000,20000000)]

c = []
import time

start = time.time()
for i in range(len(a)):
  c.append(a[i] + b[i])
print(time.time()-start)
```

```python
import sys

sys.getsizeof(a)
```

```python
import numpy as np
import sys
import time

a = np.arange(10000000)
b = np.arange(10000000,20000000)

start = time.time()
c = a + b
print(time.time()-start)
```

```python
a = np.arange(10000000,dtype=np.int8)
sys.getsizeof(a)
```

# **Fancy Indexing**
Fancy indexing is the process of selecting multiple elements, rows, or columns using a list or array of indices.

Instead of selecting one index at a time, we provide multiple index numbers together.

The outer brackets [] are used for indexing.

Inside them, NumPy expects one object.

**That object can be:**

- a single integer,
- a slice,
- or a list/array of indices

```python
# 1D
a = np.array([3, 5, 6, 7, 8, 0, 0, 0, 5])       
print(a[8])       # this is with normal indexing as we provide only one index

print()

print(a[[0, 2, 3, 6]])  # this is fancy indexing
```

```python
# 2D
a = np.array([3, 5, 6, 7, 8, 0, 0, 0, 5]).reshape(3, 3)
# selecting rows and columns
print(a)
print()
print(a[[1]])  # row
print(a[:,[2]])   # column
print(a[0, [1, 2]])
 
```

```python
# 2D
a = np.array([3, 5, 6, 7, 8, 0, 0, 0, 5]).reshape(3, 3)
print(a)
print(a[[0, 1, 2], [0, 1, 2]])               # select specific elements

print()
print(a[[0, 2], [0, 2]])  # take 0 from 0 and take 2 frome 2
print(a[[0,0,2,2], [0,2,0,2]])
```

```python
#3D

b = np.arange(1, 28).reshape(3, 3, 3)
print(b)
```

```python
print(b[[0, 2], :, :])
```

```python
print(b[[0, 0, 0, 0, 2, 2, 2, 2], [0, 0, 2, 2, 0, 0, 2, 2], [0, 2, 0, 2, 0, 2, 0, 2]])
```

# **Bolean Indexing**
Boolean indexing selects elements using True and False values.

Every True keeps the element.

Every False removes it.

array[condition]

```python
b = np.arange(1, 28).reshape(3, 3, 3)

print(b)
```

```python
a = b[b>10]*10

print(b)
print(a)
```

```python
b[b%2==0]=100
print(b)
```

```python
b[b%2!=0]*=2

print(b)
```

```python
b[(b>5) & (b<20)]
```

```python
b[(b%2==0) | (b%3==0)]
```

# **Broadcasting**

Broadcasting is NumPy's mechanism for performing element-wise operations on arrays of different shapes by automatically expanding the smaller array, without actually copying its data.

**Notice two important words:**
- Automatically expanding
- Without actually copying

Many beginners misunderstand broadcasting.

They think NumPy physically makes new copies of the smaller array.

It usually doesn't.

NumPy behaves as if the smaller array were expanded. Internally, it avoids unnecessary copies whenever possible, making broadcasting memory efficient.





### **Broadcasting Rules**

**1. Make the two arrays have the same number of dimensions.**<br>
- If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.<br>

<br>

**2. Make each dimension of the two arrays the same size.**<br>
- If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
- If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.


<img src = "https://jakevdp.github.io/PythonDataScienceHandbook/figures/02.05-broadcasting.png">

```python
a = np.arange(12).reshape(4,3)
b = np.arange(3)

print(a)
print(b)
print()
print(a+b)
```

```python
a = np.arange(12).reshape(3,4)
b = np.arange(3)

print(a)
print(b)

print(a+b)        # its does not perform broadcaste
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)

print(a)
print(b)

print(a+b)
```

```python
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)

print(a)
print(b)

print(a + b)
```

# **Working with mathematical Formula**

```python
# Sigmoid

a = np.arange(100)

def sigmoid(arr):
    return 1/(1+np.exp(-arr))

sigmoid(a)
```

```python
# Mean Square Error

a = np.array([1, 2, 3, 4, 5])
b = np.array([3, 4, 5, 6, 7])

def mse(actual, predicted):
    error = (actual - predicted)**2
    return np.mean(error)

mse(a, b)
```

# **Working with Null Values**

### **isnan**

```python
a = np.array([1, 2, 3, 4, 5, 6, 7, 8, np.nan, np.nan, 9])

print(a)
```

```python
np.isnan(a)
```

```python
# seeing null values
a[np.isnan(a)]
```

```python
# counting null values
np.isnan(a).sum()
```

```python
# seeing without null value

a[~np.isnan(a)]
```

```python
# replacing null values
a[np.isnan(a)]=0
```

```python
arr = np.array([10,20,np.nan,40])
```

```python
# replacing with mean
np.nanmean(arr)

# nanmean is used to calculate the mean by ignoring the null values
```

```python
arr[np.isnan(arr)] = np.nanmean(arr)
```

```python
print(np.round(arr, 2))
```
