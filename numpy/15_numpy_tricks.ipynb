### np.sort

Return a sorted copy of an array.

https://numpy.org/doc/stable/reference/generated/numpy.sort.html

```python
import numpy as np

arr = np.random.randint(1, 20, 25)
```

```python
np.sort(arr)
```

```python
np.sort(arr)[::-1]   # to reverse the list
```

```python
# 1D array
arr = np.random.randint(1, 20, 25).reshape(5, 5)
print(arr)
```

```python
np.sort(arr, axis=0)  # the default axis is 1
```

```python
np.sort(arr, axis=1)
```

### np.append

The numpy.append() appends values along the mentioned axis at the end of the array, it work on the same dimension array

https://numpy.org/doc/stable/reference/generated/numpy.append.html

```python
np.append(arr, arr2.reshape(1,5), axis=0)
```

```python
np.append(arr, np.arange(1, 6).reshape(5, 1), axis=1)
```

### np.where

The numpy.where() function returns the indices of elements in an input array where the given condition is satisfied.

np.where(condition, if_true, if_false )

https://numpy.org/doc/stable/reference/generated/numpy.where.html

```python
a = np.array([11, 53, 28, 50, 38, 37, 94, 92,  5, 30, 68,  9, 78,  2, 21])
a
```

```python
np.where(a>10, 0, a)
```

```python
import numpy as np
marks = np.array([45,80,30,90])

grade = np.where(
    marks>=50,
    "Pass",
    "Fail"
)

print(grade)
```

### np.ravel
convert an array into 1d just for view as it connect to the original array and it we make some changes in the converted array then it will be also apply on the original array

```python
a = np.arange(1,10).reshape(3, 3)
a
```

```python
b = np.ravel(a)
print(b)
b[0]=100
print(b)
print(a)      # notice it also change the original array
```

### obj.flatten
it create a new array and not connected with original array and work same as ravel

```python
a = np.arange(1, 10).reshape(3, 3)
print(a)

b = a.flatten()
print(b)
b[0]=1000
print(b)
print(a)
```
