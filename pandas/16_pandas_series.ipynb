```python
math = pd.Series(
    [90,80,70],
    index=["Ali","Ahmed","Sara"]
)

english = pd.Series(
    [10,15,20],
    index=["Ali","Ahmed","Sara"]
)

math+english


# Internally Pandas matches

# Ali with Ali

# Ahmed with Ahmed

# Sara with Sara
```

# **Indexing and Slicing Function**

As you start writing more Pandas code, you'll notice that experienced developers rarely write things like:

- marks[0]
- marks[2]

Instead they use 

- loc --> `Label`
- iloc --> `Index`

- #### **Use .iloc when you're thinking in terms of positions (first row, second row, last row).**
- #### **Use .loc when you're thinking in terms of labels (customer ID, student name, date, etc.).**


# **loc and iloc**
You can use loc and iloc same like above but only the difference is when you want to access the element by postion(index) then use .iloc[] and if you want to access with label then use .loc[]


| Operation            | Example                      | Result                                   |
| -------------------- | ---------------------------- | ---------------------------------------- |
| Positive indexing    | `marks.iloc[0]`              | First element                            |
| Negative indexing    | `marks.iloc[-1]`             | Last element                             |
| Label indexing       | `marks.loc["Ali"]`           | Value for label `"Ali"`                  |
| Fancy indexing       | `marks.iloc[[0,2,4]]`        | Multiple positions                       |
| Fancy label indexing | `marks.loc[["Ali","Sara"]]`  | Multiple labels                          |
| Boolean indexing     | `marks[marks>80]`            | Filter by condition                      |
| Positive slicing     | `marks.iloc[1:4]`            | Positions 1,2,3                          |
| Negative slicing     | `marks.iloc[-3:]`            | Last three positions                     |
| Label slicing        | `marks.loc["Ahmed":"Usman"]` | Includes both labels                     |

| Method              | Purpose                                    | Example                  |
| ------------------- | ------------------------------------------ | ------------------------ |
| `astype()`          | Change data type                           | `s.astype(int)`          |
| `between()`         | Check values within a range                | `s.between(10,20)`       |
| `clip()`            | Restrict values to a minimum/maximum range | `s.clip(10,50)`          |
| `duplicated()`      | Identify duplicate values                  | `s.duplicated()`         |
| `drop_duplicates()` | Remove duplicate values                    | `s.drop_duplicates()`    |
| `isnull()`          | Identify missing values                    | `s.isnull()`             |
| `dropna()`          | Remove missing values                      | `s.dropna()`             |
| `fillna()`          | Replace missing values                     | `s.fillna(0)`            |
| `isin()`            | Check membership in a list                 | `s.isin([10,20])`        |
| `apply()`           | Apply a custom function to each element    | `s.apply(lambda x: x*2)` |
| `copy()`            | Create an independent copy                 | `s.copy()`               |
