# **DataFrame**
**A DataFrame is a two-dimensional labeled data structure consisting of rows and columns.**

**Think of it as an Excel spreadsheet or a SQL table.**

| Name  | Age | Marks | City      |
| ----- | --: | ----: | --------- |
| Ali   |  20 |    85 | Peshawar  |
| Ahmed |  22 |    90 | Lahore    |
| Sara  |  21 |    75 | Karachi   |
| Usman |  23 |    95 | Islamabad |


This entire table is a DataFrame.

**In simple word:**

**A DataFrame is a two-dimensional data structure made up of multiple Series that share the same index.**

Python internally store it like this.

```
                DataFrame
        ┌────────────────────────┐
        │ Name │ Age │ Marks     │
────────┼────────────────────────┤
0       │ Ali  │ 20  │ 85        │
1       │Ahmed │ 22  │ 90        │
2       │Sara  │ 21  │ 75        │
────────┴────────────────────────┘
```

Each column is a series

```
DataFrame

       Name Series
0      Ali
1      Ahmed
2      Sara

       Age Series
0      20
1      22
2      21

      Marks Series
0      85
1      90
2      75
```

So actually DataFrame is a collection of Series
```
DataFrame
      │
      ▼
Collection of Series
```

# **Creating of DataFrame**

### **creating from Dictionary**
By creating dataframe from dictionary each key become column of dataframe and each value(list) of dictionary become values of column

````python

data = {
    "Name":["Ali","Ahmed","Sara"],
    "Age":[20,22,21],
    "Marks":[85,90,75]
}

df = pd.DataFrame(data)

df
````

### **creting from Dictionary of Series**

````python
names = pd.Series(["Ali","Ahmed","Sara"])

ages = pd.Series([20,22,21])

marks = pd.Series([85,90,75])

dic = {
    "Name":names, 
    "Age":ages,
    "Marks":marks
}

df = pd.DataFrame(dic)

df
````

### **creating with 2D List**
When you created any Dataframe from list you have to mentioned the name of columns in the dataframe. Without mentioning the columns it will give index to the column by default.

````python
li = [
    ["salman", 34, 45, 67], 
    ["naveed", 32, 54, 65],
    ["ubaid", 34, 54, 76]
]

df = pd.DataFrame(li, columns=["Names", "Math", "Pak", "Science"])
df
````

````python
# We can also change the index and label the index of dataframe

li = [
    ["salman", 34, 45, 67], 
    ["naveed", 32, 54, 65],
    ["ubaid", 34, 54, 76]
]

df = pd.DataFrame(li, columns=["Names", "Math", "Pak", "Science"], index=["A", "B", "C"])
df
````

````python
# We can also change the index with column 
li = [
    ["salman", 34, 45, 67], 
    ["naveed", 32, 54, 65],
    ["ubaid", 34, 54, 76]
]

df = pd.DataFrame(li, columns=["Names", "Math", "Pak", "Science"])
df.set_index("Names")
````

````python
# we can also reset the index with its defualt index and the column will again become column(series)
df.reset_index()
````

# **How to reduce the memory usage of dataframe**
We can reduce the memory usage of any dataframe by change its datatype.

````python
movies.head(2)
````

````python
movies.info()    # now check the memory usage it use 2.0 + MB so we have to reduce it
````

**Look in the above info carefully :**

- The memory usage is 2.0+ MB
  
- `imdb_rating`  datatye is float64 which use high memory so we have to convert it into float32 because float32 is made for less range digits and use less memory.
  
- `is_adult` & `year_of_release` have datatypes int64 and we know this both are act like a category datatypes because 0 and 1 repeated in the `is_adult` and years repeated in `year_of_release` so we can convert it into category datatype to reduce the memory

- In the same way we can change some other datatypes to reduce its memory usage.

````python
movies['imdb_rating'] = movies['imdb_rating'].astype('Float32')
````

````python
movies.info()
````

````python
movies[["year_of_release", "is_adult"]] = movies[["year_of_release", "is_adult"]].astype("category")
````

````python
movies.info()
````

### **Notice here the + sign from 2.0 + MB remove and it convert to 2.0 MB**
It means that the memory of dataframe is reducing. Actully the movies dataframe is to much big and we cannot see the difference as it is in MB
