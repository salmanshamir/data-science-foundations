# **Groupby**
A GroupBy Object is a special Pandas object created by the groupby() method.

It does not immediately perform any calculations.

Instead, it:

- Splits the data into groups.
- Stores information about those groups.
- Waits for you to tell it what operation to perform.

Think of it as a manager that organizes your data but doesn't calculate anything until you ask.

# **The Big Idea**

**groupby() follows a famous three-step process:**

```
Split
   ↓
Apply
   ↓
Combine
```

This is known as the Split–Apply–Combine strategy.

- **Split:** Divide data into groups.
- **Apply:** Perform a function on each group.
- **Combine:** Merge the results into a single object.
  
**There should be always one categorical column for grouping and one or more numerical columns for applying any operation**

````python
# import libraries
import numpy as np
import pandas as pd
````

````python
# load the datasets
ipl = pd.read_csv('data/19_data/deliveries.csv')
movies = pd.read_csv('data/19_data/imdb-top-1000.csv')
````

````python
# making dataframe to explain the groupby object

df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Bilal", "Hassan", "Ayesha"],
    "Department": ["IT", "HR", "IT", "HR", "Sales", "IT"],
    "Salary": [50000, 60000, 70000, 65000, 45000, 55000]
})
````

````python
# it actually give a groupby object

grouped = df.groupby("Department")
print(grouped)        # it is a dataframegroupby object
type(grouped)
````

````python
# we can perform any operation on groupby object

grouped.sum()
````

````python
# we can select any column after applying any operation
grouped.sum()["Salary"]
````

````python
# we can also do the same process in this way
grouped["Salary"].sum()          # this method is much faster than the above because the above method first apply operation on all columns
                                # and then we select columns while this process first select column then perform operation on selected column
````

````python
# we can apply any math funtion as
grouped["Salary"].min()
````

````python
grouped["Salary"].max()
````

````python
grouped["Salary"].std()
````

# **Groupby Attribute**

### **groups**
groups returns a dictionary where:

- Key = Group name
- Value = Index labels belonging to that group not the actual index position of row.

````python
grouped.groups
````

````python
ipl.set_index('batting_team').groupby('batsman').groups  # as it will return the dic with values of labels
````

### **indices**
indices returns a dictionary where

```
Key

↓

Group name

Value

↓

Integer positions of rows. not label
```

````python
grouped.indices
````

````python
ipl.set_index('batting_team').groupby('batsman').indices # it will instead return the index position
````

### **ngroups**
Returns the total number of groups.

````python
grouped.ngroups
````

````python
# similarly
len(grouped)
````

# **Summary**
| Attribute | Return Type | What It Returns                                 | Example Output                                  |
| --------- | ----------- | ----------------------------------------------- | ----------------------------------------------- |
| `groups`  | `dict`      | Group names mapped to **index labels**          | `{'IT': Index([0, 2, 5]), 'HR': Index([1, 3])}` |
| `indices` | `dict`      | Group names mapped to **integer row positions** | `{'IT': array([0, 2, 5]), 'HR': array([1, 3])}` |
| `ngroups` | `int`       | Number of groups                                | `3`                                             |

# **Groupby Methods**

### **size()**
Returns the number of rows in each group.

It counts every row, including rows containing NaN.

**syntax:**

grouped.size()

````python
grouped.size()
````

````python
# similary

df["Department"].value_counts()
````

### **get_group()**

get_group() returns one complete group from the GroupBy object.

Instead of returning all groups, it returns only the group you ask for.

**syntax:**

grouped.get_group(name)

````python
grouped.get_group('HR')
````

````python
grouped.get_group("Sales")
````

````python
grouped.get_group("IT")
````

### **first()**

Returns the first non-null row/value from each group.

If the first row has `NaN` in one column, Pandas may use the next non-null value for that specific column.

**syntax:**

grouped.first()

````python
grouped.first()
````

### **last()**

Exactly opposite of first().

Returns the last non-null row/value from every group.

**syntax:**

grouped.last()

````python
grouped.last()
````

### **sample()**

It return any random rows from each group

**syntax:**

grouped.sample()

````python
grouped.sample()
````

````python
grouped.sample(2, replace=True)
````

### **nth()**

This is more flexible than first() and last().

Returns the nth row from every group.

**syntax:**

grouped.nth(n)

| Parameter | Default  | Accepted Values                | Purpose                                    | Explanation                                                                 |
| --------- | -------- | ------------------------------ | ------------------------------------------ | --------------------------------------------------------------------------- |
| `n`       | Required | Integer (positive or negative) | Choose which row to return from each group | `0` = first row, `1` = second row, `-1` = last row, `-2` = second last row. |


````python
grouped.nth(0)      # give the first as grouped.first()
````

````python
# similary 
grouped.first()
````

````python
# but we can get any row number of each group

grouped.nth(2)          # it give only the IT because IT group only contain 3 rows
````

### **describe()**

Returns a complete statistical summary for every group.

**syntax:**

grouped.describe()

````python
grouped.describe()
````

### **nunique()**

Counts the number of unique values inside every group.

**syntax:**

grouped.nunique()

````python
grouped.nunique()
````

### **agg()**

The agg() method (short for aggregate) applies one or more aggregation functions to a Series, DataFrame, or GroupBy object and combines the results into a single output.

**syntax:**

grouped.agg(fun)

````python
grouped.agg('min')
````

````python
grouped.agg('max')
````

````python
# on single colunns
grouped["Salary"].mean()
````

````python
# multiple operation
grouped.agg(['min', 'max'])
````

````python
grouped["Salary"].agg(["min", "max"])
````

````python
# specific operation on specific columns
grouped.agg(
    {
        "Salary":'mean',
            "Name":'min'
    }
)
````

````python
grouped.agg({
    "Salary":["min", "max", "mean"],
    "Name":["min", "max"]
})          
# actually this is a multiindex dataframe
````

````python
# giving name to each colunmn 

grouped.agg(
    salary_min=('Salary', "min"),
    salary_max=('Salary', "max")
)
# this is not a multiindex dataframe its a simple dataframe
````

### **apply()**
GroupBy.apply() applies a user-defined function (or a built-in function) to each group independently, then combines the results into one final object.

In one sentence:

Split the data → Send each group to your function → Collect every result → Return one combined object.

This is why it is called apply.



````python
# it will return the min of salary of each group 
def mini(group):
    return group["Salary"].min()

grouped.apply(mini)       # in this way we can make any custom function and apply on the grouped object
````

````python
def maxi(group):
    return {
        "Salary_max":group["Salary"].max(),
        "Salary_min":group["Salary"].min()
    }
    
grouped.apply(maxi)
````

````python
def statistics(group):

    return pd.Series({

        "Average":group["Salary"].mean(),

        "Max":group["Salary"].max(),

        "Min":group["Salary"].min()

    })

grouped.apply(statistics)
````

````python
grouped.apply(lambda x: x.nlargest(1, 'Salary'))
````

# **Looping through groupby object**

We can iterate through groupby object and it give a tuple with two values the first is the name of group and second is dataframe or series.

````python
for name,data in grouped:
    print(type(name))
    print(type(data))
````

````python
for name,data in grouped:
    print(name.center(30))
    print(data)
    print("-"*30)
    
````

# **Some Question with Groupby**

### **find number of movies starting with A for each group**

````python

grouped = movies.groupby("Genre")

grouped.apply(lambda x: x[x["Series_Title"].str.startswith("A")])
````

### **find ranking of each movie in the group according to IMDB score**

````python
def ran(group):
    group["Rank"] = group["IMDB_Rating"].rank(ascending=False)
    return group
````

````python
grouped.apply(ran)
````

### **find normalized IMDB rating group wise**

````python
def normal(group):
  group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group
````

````python
grouped.apply(normal).reset_index(level=0)
````

### **find the most earning actor->director combo**

````python
movies.groupby(["Director", "Star1"])["Gross"].sum().sort_values(ascending=False).head(1)
````

### **find the best(in-terms of metascore(avg)) actor->genre combo**

````python
group = movies.groupby(["Genre", "Star1"])["Metascore"].mean().reset_index().sort_values("Metascore", ascending=False)
group.nlargest(1, "Metascore", keep="all")
````

### **find the top 10 batsman in terms of runs**

````python
ipl.groupby("batsman")["batsman_runs"].sum().sort_values(ascending=False).head(10)
````

### **find the batsman with max no of sixes**

````python
six = ipl[ipl["batsman_runs"]==6]

six.groupby("batsman")["batsman"].count().sort_values(ascending=False).head(1).index[0]
````

### **find batsman with most number of 4's and 6's in last 5 overs**

````python
last_over = ipl[ipl["over"]>15]
last_over[last_over["batsman_runs"].isin([6, 4])].groupby("batsman")["batsman_runs"].sum().sort_values(ascending=False).head(1).index[0]
````
