### `Q-2:` Using `groupby` make groups using the `"Pclass"` column and fill every group's `"Embarked"` column's missing values with the mode value of that group. After that, print every group's `"Embarked"` column's value counts in ascending order.

````python
# code here

grouped = titanic.groupby("Pclass")["Embarked"].transform(lambda x:x.fillna(x.mode().iloc[0]))

grouped = titanic.groupby("Pclass")["Embarked"]

for name, data in grouped:
    print(name)
    print(data.value_counts().sort_values())
    print("*"*30)
````

```bash
{'C': {1: 105, 2: 25, 3: 11},
```

### `Q-4:` Perform `groupby` based on the `"Team"` column and then perform Z Normalization on top of the below columns of each group:
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks

You have to make a python function named `z_normalization` which takes two arguments:

1. *group:* Every group that you have created
2. *cols_to_perform:* This parameter takes a list of columns on which you have to perform the Z-Normalization.

For this problem, you have to use th `apply()` method.

$\Large Z \ - \ Normalization = \frac{X_i - \mu}{std} $


````python
# code here
pd.options.display.max_columns = None
fifa = pd.read_csv("data/19_data/Fifa Worldcup 2022.csv")
grouped = fifa.groupby("Team")

def z_normalize(group, li):
    for i in li:
        group[i] = (group[i]-group[i].mean())/np.std(group[i])
    return group

l = ["Passes", "Passes Completed", "Attempted Line Breaks", "Completed Line Breaks"]


grouped.apply(z_normalize, l)        
    
        
        
````

### `Q-5:` Find batsman in below category-
* Highest score while chasing
* Best Strike rate while chasing and have faced 100+ balls


> Chasing mean team batting in second inning

````python
# code here
ipl = pd.read_csv('data/19_data/deliveries.csv')
fil = ipl[ipl["inning"]==2]
fil.groupby(["match_id", "batsman"])["batsman_runs"].sum().sort_values(ascending=False)
st = fil.groupby(["match_id", "batsman"])["batsman_runs"].agg(['sum', 'count'])
st = st.reset_index()
st.rename(columns={
    "sum":'runs', 
    "count":"ball"
}, inplace=True)
st_100 = st[st["ball"]>100]
st_100["rate"] = (st_100["runs"]/st_100["ball"])*100
st_100.sort_values(by="rate", ascending=False)
````

### `Q-8:` Make a dataframe for all batting pairs played together.
```
Batsman1 Batsman2 Runs Avg StrikeRate
```

> Just to ease this question you can count wide-balls for strike rate.

````python
# code here


ipl[["batsman1", 'batsman2']] = ipl["Pair"].str.split(" - ", expand=True)
fil = ipl.groupby(["batsman1", "batsman2"])[["wide_runs", "batsman_runs"]].sum().reset_index()
fil["total_runs"] = fil["wide_runs"] + fil["batsman_runs"]
fil.drop(columns=["wide_runs", "batsman_runs"], inplace=True)
````

````python
dff = ipl.groupby(["batsman1", "batsman2"])["batsman_runs"].count().sort_values(ascending=False).reset_index()
````

````python
dff.rename(columns={
    "batsman_runs":"balls"
}, inplace=True)
````

````python
final = pd.merge(fil, dff, on=["batsman1", "batsman2"], how='inner')
final["strike_rate"] = round((final["total_runs"]/final["balls"])*100, 2)
````
