<b><h1><center>Python Programming Language</center></h1></b>

<b><h3>What is Python?</h3></b>
<p>Python is a high-level, interpreted, general-purpose programming language that is known for its simple and readable syntax. It allows developers to write clean, efficient code and is widely used in web development, data science, machine learning, artificial intelligence, automation, scientific computing, and many other fields.</p>

<br>

<b><h3>History of Python</h3></b>
<p>Python was created by Guido van Rossum in 1989 at the Centrum Wiskunde & Informatica (CWI) in the Netherlands.<br>
The first public release of Python (Version 0.9.0) was made available in 1991.<br>
Guido designed Python with the goal of making programming more readable and enjoyable by emphasizing simplicity and code readability.</p>

<br>

<b><h3>Why is it called "Python"?</h3></b>
<p>Despite what many beginners think, Python is not named after the snake.<br>
Guido van Rossum named the language after the British comedy television show "Monty Python's Flying Circus", because he wanted the language to have a short, unique, and memorable name.</p>

<br>

<b>Large Standard Library</b> – Python includes many built-in modules that simplify programming tasks.<br>
<b>Extensive Community Support</b> – Millions of developers contribute tutorials, libraries, and solutions.<br>
<b>Versatile</b> – Python can be used for web development, automation, data analysis, AI, game development, scripting, and more.</p>

<b><h1><center>Print Function</center></h1></b>

<b><h3>Definition</h3></b>
<p>The print() function is a built-in Python function used to display output on the screen. It allows programmers to print text, numbers, variables, expressions, and the results of calculations.<br>
It is one of the first and most frequently used functions in Python because it helps us display information and debug programs.</p>

<br>

<b><h3>Parameters</h3></b>
<p>object(s): The value(s) to be displayed. You can pass strings, numbers, variables, or expressions.<br>
sep (optional): Specifies the separator between multiple objects. The default separator is a space (' ').<br>
end (optional): Specifies what to print at the end of the output. The default is a newline ('\n').</p>

```python
print("Hello World")   # Out first program
```

```python
print("salman")
```

```python
print("muhammad")
print("salman")
```

```python
print("muhammad", end=" ")
print("salman")
```

```python
print("muhammad", "salman", sep="-")
```

```python
print(8, True, "salman", sep="/")
```

<b><h3>1. Integer(int)</h3></b>

<b><h3>6. String (str)</h3></b>
<p>A string is a sequence of characters enclosed in single quotes (' ') or double quotes (" ").</p>

```python
# String with double quotation
print("muhammad")  

# String with single quotation
print('Salman')    

# Multiple lines comment
print("""Hi!
How are YOU.
Hope you are doing well.
      """)        
```

<b><h3>7. List (list)</h3></b>
<p>A list is an ordered and mutable collection of items. Lists are created using square brackets [ ].</p>

```python
print([12, 34, 3, 34])

# We can store multiple datatypes in a list
print(["salman", 3, True, None, 3,4])
```

<b><h3>8. Tuple (tuple)</h3></b>
<p>A tuple is an ordered but immutable collection of items. Tuples are created using parentheses ( ). This is like list but it is immutable as we cannot change the items in a tuple.</p>

```python
print((23, 3, 45, 3))

# we can also store multiple datatypes in a tuples.
print((2, 2.3, True, None, "salman"))
```

<b><h3>9. Set (set)</h3></b>
<p>A set is an unordered collection of unique elements. Sets are created using curly braces { }. We cannot store duplicates values in a set as set ignore the duplicates value.</p>

```python
# The order of the values in a set changes 
print({3, 34, 32, 5, 9})

# It ignore the duplicates
print({23, 23, 24})

# We can also store multiple datatypes in a set
print({3, 3.4, True, None, "salman"})
```

<b><h3>10. Dictionary (dict)</h3></b>
<p>A dictionary stores data in key-value pairs. Dictionaries are created using curly braces { }.</p>

```python
print({
    "key":"value", 
    "Name":"Salman",
    "Marks":45
})
```

<b><h3>Checking the Data Type</h3></b>
<p>Python provides the type() function to determine the data type of any object.</p>

```python
print(type(34))    # integer
print(type(4.5))   # float
print(type(3+6j))   # complex
print(type(True))   # Boolean
print(type(False))   # Boolean
print(type(None))    # None
print(type("salman")) # string
print(type([3, 2, "salman", 4,5]))  # list
print(type((2, 1, "salman")))   # tuple
print(type({3, 2, 5}))       # set
print(type({"Name": "Salman"}))   # dictionary
```

<b><h1><center>Variables in Python</center></h1></b>

<b><h3>Variables</h3></b>
<p>Variables is like a container that holds a value, which can be used, modified, or accessed throughout the program. In Python, a variable is created automatically when you assign a value to it. You do not need to declare its data type explicitly because Python is a dynamically typed language.</p>

<br>

<b><h3>Variable Naming Rules</h3></b>
<b><p>When naming variables in Python, follow these rules:</b><br>
<b>--></b> Variable names can contain letters, digits, and underscores (_).<br>
<b>--></b> A variable name must start with a letter or an underscore.<br>
<b>--></b> A variable name cannot start with a digit.<br>
<b>--></b> Variable names are case-sensitive (age and Age are different variables).<br>
<b>--></b> Variable names cannot be Python keywords.</p>

```python
name = "salman"
age = 19
print(name)
print(age)
```

<b><h3>Multiple Variable Assignment</h3></b>
<p>Python allows assigning values to multiple variables in a single line.</p>

```python
x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
```

<b><h3>Assigning the Same Value to Multiple Variables</h3></b>
<p>You can assign one value to multiple variables.</p>

```python
a = b = c = 10
print(a)
print(b)
print(c)
```

<b><h3>Changing the Value of a Variable</h3></b>
<p>A variable's value can be changed at any time.</p>

```python
marks = 34
print(marks)

marks = 90
print(marks)
```

<b><h3>Static Typing</h3></b>
<p>Static typing is a programming approach in which the data type of a variable is checked before the program runs (at compile time). Once a variable is declared with a specific data type, it cannot store a value of another data type unless it is explicitly converted.

Languages like C, C++, Java, and C# are statically typed.</p>

```python
# This is java code

# int age = 18;
# age = 20;      // ✅ Correct

# age = "Salman";  // ❌ Error
```

<b><h3>Dynamic Typing</h3></b>
<p>Dynamic typing is a programming approach in which the data type of a variable is determined automatically at runtime. You do not need to declare the variable's type.

Python is a dynamically typed language.</p>

```python
value = 90
print(value)
print(type(value))

# storing string in the same variable
value = "salman"
print(value)
print(type(value))
```

| Static Typing                            | Dynamic Typing                                            |
| ---------------------------------------- | --------------------------------------------------------- |
| Data type is declared by the programmer. | Data type is determined automatically by the language.    |
| Checked before execution (compile time). | Checked during execution (runtime).                       |
| Less flexible.                           | More flexible.                                            |
| Faster execution.                        | Slightly slower because type checking happens at runtime. |
| Example: Java, C, C++                    | Example: Python, JavaScript, Ruby                         |

<b><h1><center>Comments in Python</center></h1></b>

<b><h3>Comments</h3></b>
<p>Comments are lines of text in a Python program that are ignored by the Python interpreter during execution. They are used to explain the purpose of the code, improve readability, and make the program easier to understand and maintain.</p>

<b><h3>1. Single-Line Comment</h3></b>
<p>A single-line comment starts with the # symbol.</p>


```python
# this is a single line comment and python completely ignore this line while executing the program
print("Hi! How are you")
```

<b><h3>1. Multi-Line Comment</h3></b>
<p>Python does not have a dedicated syntax for multi-line comments. However, programmers commonly use triple quotes (''' or """) to write multi-line descriptions.</p>

<b>Note:</b> Triple-quoted strings are technically string literals. When they are not assigned to a variable, they are often used as multi-line comments or documentation.


```python
"""This is a multiple
line comment and python will ignore 
these all lines"""

print("how are you")
```

<b><h1><center>Type Conversion</center></h1></b>

<b><h3>Type Conversion</h3></b>
<p>Type conversion is the process of converting a value from one data type to another.

<b>Python supports two types of type conversion:</b><br>
1. Implicit Type Conversion<br>
2. Explicit Type Conversion</p>


<b><h3>1. Implicit Type Conversion</h3></b>
<p>Python automatically converts one data type into another when it is safe to do so.</p>

```python
num_1 = 2    # integer
num_2 = 2.2   # float
sum = num_1 + num_2
print(sum)
print(type(sum))  # become float
```

<b><h3>2. Explicit Type Conversion</h3></b>
<p>The programmer manually converts one data type into another using built-in functions.</p>

```python
num_1 = 1.3
print(type(num_1))

num_1 = int(num_1)
print(type(num_1))
```
