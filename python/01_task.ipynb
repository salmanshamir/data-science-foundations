### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python

# lets try this problem first with algebra

# let suppos
# dogs = d
# chicken = c

# as we know total dogs and chicken is equal to the heads of dogs and chicken because every dog and chicken have on heads
# d + c = heads
# c = heads - d     this is for chicken  by adjustanting the above equaton


# what about d

# legs = 2d + 4c        put heads - d in place of c
# legs = 2d + 4(heads-d)
# legs = 2d + 4heads - 4d
# legs = -2d + 4heads
# legs-4heads=-2d
# -2d = legs -4heads
# -d = legs-4heads/2
# d = -(legs-4heads)/2




# Write your code here

heads=int(input("enter heads"))
legs=int(input("enter legs"))


dogs = (-legs+4*heads)//2

chickens = heads-dogs

print("number of dogs", dogs)
print("number of chickens", chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here

num = int(input("enter number : "))

sum = 0
for i in range(1, num+1):
    square = i**2
    sum = sum+square
    
print(sum)
````
