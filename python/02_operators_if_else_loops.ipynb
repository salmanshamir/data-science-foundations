<b><h1><center>Operators in Python</center></h1></b>

<b><h3>Operators</h3></b>
<p>Operators are special symbols or keywords that perform operations on one or more operands (values or variables) to produce a result.</p>

```python
a = 12
b = 10
print(a+b)
```

<b>Here:</b><br>
<p>a and b are operands.<br>
+ is the operator.</p>

### **Types of Operators**

<p>Python provides the following types of operators:</p>

| Operator Type        | Description                          |
| -------------------- | ------------------------------------ |
| Arithmetic Operators | Perform mathematical operations      |
| Assignment Operators | Assign values to variables           |
| Comparison Operators | Compare two values                   |
| Logical Operators    | Combine multiple conditions          |
| Identity Operators   | Compare object identity              |
| Membership Operators | Check membership in a sequence       |
| Bitwise Operators    | Perform operations on binary numbers |




<br>

### **1. Arithmetic Operators**

Used for mathematical calculations.

| Operator | Meaning        | Example  | Result  |
| -------- | -------------- | -------- | ------- |
| `+`      | Addition       | `5 + 3`  | `8`  |
| `-`      | Subtraction    | `5 - 3`  | `2`  |
| `*`      | Multiplication | `5 * 3`  | `15`  |
| `/`      | Division       | `5 / 2`  | `2.5`  |
| `//`     | Floor Division | `5 // 2` | `2`  |
| `%`      | Modulus        | `5 % 2`  | `1`  |
| `**`     | Exponent       | `5 ** 2` | `25`  |


```python
a = 5
b = 3

print(a+b)
print(a-b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(5 ** 2)
```

### **2. Assignment Operators**
Assign values to variables.

| Operator | Example        | Meaning  |
| -------- | -------------- | -------- |
|   =      | x = 10       | x = 10  |
|   +=     | x += 9    | x = x + 9  |
|   -=     | x -= 3 | x = x - 3  |
|    *=    | x *= 5       | x = x * 5  | 
|    /=    | x /= 2 | x = x / 2 |
|    //=    | x //= 4 | x = x // 4 |

```python
x = 10
print(x)

x += 10
print(x)

x -= 10
print(x)

x *= 10
print(x)

x /= 10
print(x)

x //= 3
print(x)


```

### **3. Comparison Operators**
Return either True or False.

| Operator | Meaning                  |
| -------- | ------------------------ |
| `==`     | Equal to                 |
| `!=`     | Not equal to             |
| `>`      | Greater than             |
| `<`      | Less than                |
| `>=`     | Greater than or equal to |
| `<=`     | Less than or equal to    |


```python
a = 5
b = 10
c = 10

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(b >= c)
print(c <= b)
```

### **4. Logical Operators**
Used to combine multiple conditions and return either True or False.

| Operator | Meaning                             |
| -------- | ----------------------------------- |
| `and`    | Both conditions must be True        |
| `or`     | At least one condition must be True |
| `not`    | Reverses the result                 |


```python
a = 19
b = 10
c = 10
print(a > 10 and b > 20)   # False because b is not greater 20

print(a > 10 or b > 20)   # True because a is greater than 10

print(not(a > 10 or b > 20))  # False because it reverse the result
```

### **5. Identity Operators**
Compare whether two variables refer to the same object.

**is**<br>
**is not**

```python
a = "salman"
b = "muhamad"
c = "salman"
d = 10
e = 10

print(a is b)
print(d is e)
print(a is c)
print(a is not c)
```

### **6. Membership Operators**

Check whether a value exists in a sequence.

**in**<br>
**not in**

```python
a = "salman"
b = ["apple", "orange", "banana"]

print("a" in a)
print("a" not in a)
print("apple" in b)
print("orange" not in b)
print("patato" in b)
```

<b><h1><center>If-Else Statement</center></h1></b>

### **1. if Statement**
Executes a block of code only if the condition is True.

```python
age = 20

if age >= 18:
    print("You are eligible")
```

### **2. if-else Statement***

```python
age = 10

if age >= 18:
    print("You are eligible")
else:
    print("You are not eligible")
```

### **3. if-elif-else Statement**

```python
marks = 30

if marks >= 80:
    print("Grade-A")
elif marks >= 60:
    print("Grade-B")
elif marks >= 40:
    print("Grade-C")
else:
    print("Grade-D")
```

### **4. Nested if**

An if statement inside another if statement is called a nested if.

```python
age = 100
citizen = True

if age >= 18:
    if citizen:
        print("You are eligible")
    else:
        print("You are not eligible")
else:
    print("You are not eligible")
```

<b><h1><center>Loops in Python</center></h1></b>

### **Loop**
A loop repeatedly executes a block of code until a condition becomes False or the sequence is exhausted.

<br>

### **Why Do We Use Loops?**
Loops reduce repetition and make code cleaner.

**Without loops:**

```python
print("Hello World")
print("Hello World")
print("Hello World")
```

**With loops:**

```python
for i in range(1, 4):
    print("Hello World")
```


### **1. For Loop**
Used when the number of iterations is known. <br>

For loop take three attribute:<br>

**start**<br>
**stop**<br>
**step**<br>

```python
# 1 will be included and 5 will be excluded
for i in range(1, 5):
    print(i)
    
```

```python
for i in range(1, 5, 2):
    print(i)
```

### **2. While Loop**

Executes while a condition remains True.

```python
count = 1

while count < 5:
    print(count)
    count+=1
```

### **Difference Between for and while**

| for Loop                       | while Loop                       |
| ------------------------------ | -------------------------------- |
| Used when iterations are known | Used when iterations are unknown |
| Iterates over a sequence       | Depends on a condition           |

<br>
<br>

### **Nested Loop**
A loop inside another loop.

```python
for i in range(1, 3):
    for j in range(1, 5):
        print(i, j)
```

<b><h1><center>Control Statements in Python</center></h1></b>

### **Control Statements**
Control statements change the normal execution of loops.

### **1. Break Statement**
The break statement immediately terminates the nearest enclosing loop.

```python
for i in range(1, 11):
    print(i)
    if i==5:
        break
```

### **2. continue Statement**
The continue statement skips the current iteration and moves to the next iteration of the loop.

```python
for i in range(1, 6):
    if i==3:
        continue
    print(i)
```

### 7 - Write a program to keep accepting numbers from users till he/she enters a 0 and then find the avg

```python
sum = 0
count = 0



while True:
    num = int(input("Enter number : "))
    
    if num == 0:
        break
    else:
        sum += num
        count+=1
    
if sum == 0:
    avg = "Nothing to calculate average"
else:
    avg = sum/count

print(avg)
```

### 9 - Write a program to calculate the sequence sum 1/1! + 2/2! + 3/3! + ...n

```python
n = int(input("Enter n number  : "))
fact = 1
result = 0
for i in range(1, n+1):
    fact *= i
    result = result + i/fact
    
print(result)
```
