### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
salary_ctc = int(input("enter your salary with ctc"))

hr = salary_ctc*10/100
da = salary_ctc*5/100
pf = salary_ctc*3/100

deduction_amount = hr+da+pf

salary_with_tax = salary_ctc-deduction_amount

if salary_ctc<500000:
    salary = salary_with_tax
elif salary_ctc>=500000 and salary_ctc<1000000:
    salary = salary_with_tax-(salary_ctc*10/100)
elif salary_ctc>=1000000 and salary_ctc<2000000:
    salary = salary_with_tax-(salary_ctc*20/100)
else:
    salary = salary_with_tax-(salary_ctc*30/100)


print(salary)    
    
    
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here

num = int(input("enter number"))

factorials = 1 
for i in range(1, num+1):
    factorials*=i
    
print(factorials)
````

###`Problem 11`: A robot moves in a plane starting from the original point (0,0). The robot can move toward UP, DOWN, LEFT and RIGHT with a given steps.
The trace of robot movement is shown as the following:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
> The numbers after the direction are steps.

> `!` means robot stop there.

**Please write a program to compute the distance from current position after a sequence of movement and original point.**

*If the distance is a float, then just print the nearest integer.*

Example:

`Input`:
```
UP 5
DOWN 3
LEFT 3
RIGHT 2
!
```
`Output`:
```
2
```

````python
# Write code here

import math

x = 0

y = 0
while True:
    mov = input("""
                    1. UP
                    2. DOWN
                    3. LEFT
                    4. RIGHT
                    !. STOP
                    """)
    
    
    
    if mov=="1":
        up = int(input("Enter up steps"))
        y = y+up
    elif mov=="2":
        down = int(input("enter down steps"))
        y = y-down
    elif mov=="3":
        left = int(input("enter left steps"))
        x = x-left
    elif mov=="4":
        right = int(input("enter right steps"))
        x = x+right
    elif mov=="!":
        break
    else:
        print("invalid option")
        

position = f"({x}, {y})"

distance = math.sqrt((x)**2+(y)**2)

print(f"The ROBOT Successfully stop at {position} position.")
print()
print(f"The distance of robot from origin {int(distance)} ")
        

    
        

````

### `Problem 12`:Write a program to print whether a given number is a prime number or not

````python
# Write code here

num = int(input("enter any no"))

status = True
for i in range(1, num+1):
    if i!=1 and i!=num:
        if num%i==0:
            status=False
            break
    else:
        status=True
        
if status==True:
    print("prime")
else:
    print("Not prime")
````
