# **String**
A string is a sequence of characters enclosed within single quotes (' '), double quotes (" "), or triple quotes (''' ''' or """ """).

A string can contain letters, numbers, symbols, spaces, and even special characters.

In Python, strings are immutable, meaning once a string is created, its characters cannot be changed directly.


<br>

# **Creating String**

There are three common ways to create strings.

### **1. Single Quotes**


```python
lan = 'Python'
print(lan)
```

### **1. Double Quotes**

```python
lan = "Python"
print(lan)
```

### **1. Triple Quotes**
Used for multi-line strings.

```python
lan = """Python 
is a power 
language"""

print(lan)
```

# **String Indexing**

Each character in a string has an index.

<span>
Character :  P  y  t  h  o  n   <br>
Positive :   0  1  2  3  4  5   <br>
Negative :  -6 -5 -4 -3 -2 -1   <br>
</span>

### **1. Positive Indexing**

```python
a = "python"
print(a[0])
print(a[1])
print(a[2])
print(a[3])
print(a[4])
print(a[5])
```

### **2. Negative Indexing**

```python
a = "python"
print(a[-1])
print(a[-2])
print(a[-3])
print(a[-4])
print(a[-5])
print(a[-6])
```

# **String Slicing**
string[start : stop : step]

```python
a = "python"
print(a[0:5])  # 0 included and 5 excluded
print(a[1:6])
print(a[0:6:2])
print(a[-5:])
print(a[-5:1])
```

### **Reversing a String**

```python
a = "python"
print(a[::-1])
```

# **Editing Strings**

# **Some Function and Methods of String**

### **1. len()**
The len() function returns the total number of characters (length) in a string.

```python
a = "python"
print(len(a))
```

### **2. max()**
Returns the character with the highest Unicode (ASCII) value.

```python
a = "salman"
print(max(a))
```

### **3. min()**
Returns the character with the smallest Unicode value.

```python
a = "salman"
print(min(a))
```

### **4. sorted()**
Returns the characters of a string in sorted order.

```python
a = "salman"
print(sorted(a))
```

### **5. ord()**
Returns the Unicode value of a single character.

```python
print(ord("a"))
print(ord("A"))
print(ord("b"))
print(ord("q"))
```

### **6. chr()**
Returns the character corresponding to a Unicode value.

```python
print(chr(97))
print(chr(65))
print(chr(113))
```

### **7. upper()**
Converts all characters to uppercase.

```python
a = "salman"
print(a.upper())
```

### **8. lower()**
Converts all characters to lowercase.

```python
a = "salman"
print(a.lower())
```

### **9. strip()**
Removes whitespace (or specified characters) from both ends of a string.

```python
a = "              salman           "
print(a.strip())
```

### **10. replace()**
Replaces occurrences of a substring with another substring.<br>

**Syntax:**
string.replace(old, new, count)

| Parameter | Description                     |
| --------- | ------------------------------- |
| old       | Text to replace                 |
| new       | New text                        |
| count     | Optional number of replacements |


```python
a = "salman"
print(a.replace("a", "k"))
print(a.replace("a", "b", 1))
b = "muhammad salman"
print(b.replace("salman", "naveed"))
```

### **11. split()**
Splits a string into a list.

**Syntax:**<br>
string.split(separator, maxsplit)

| Parameter | Description               |
| --------- | ------------------------- |
| separator | Delimiter (optional)      |
| maxsplit  | Maximum splits (optional) |


```python
a = "muhammad salman-muhammad naveed-muhammad junaid"
print(a.split())
print(a.split(sep="-"))
print(a.split(sep="-", maxsplit=1)) 
print(a.split("-", 2))
```

### **12. join()**
Joins elements of an iterable into one string.

**Syntax:**<br>
separator.join(iterable)

| Parameter | Description       |
| --------- | ----------------- |
| iterable  | List, tuple, etc. |


```python
a = ["salman", "naveed", "junaid"]
print(" ".join(a))
print("-".join(a))
print(" ".join(a[1:]))
```

### **13. find()**
Returns the index of the first occurrence of a substring.<br>

**Syntax:**<br>
string.find(sub, start, end)

```python
a = "salman"
print(a.find("a"))  # for first a
print(a.find("a",2,)) # for second a
print(a.find("k"))    # return -1 if the character is not present
```

### **14. count()**
Counts the occurrences of a substring.<br>

**Syntax:**<br>
string.count(sub, start, end)



```python
a = "muhammad salman"
print(a.count("a"))
print(a.count("a", 0, 8)) # only for muhammad 
```

### **15. startswith()**
Checks whether a string starts with the specified prefix.

```python
a = "salman"
print(a.startswith("s"))
print(a.startswith("sa"))
print(a.startswith("b"))
```

### **16. endswith()**
Checks whether a string ends with the specified suffix.

```python
a = "salman"
print(a.endswith("an"))
print(a.endswith("ma"))
```

### **17. isalpha()**
Returns True if all characters are alphabets

```python
a = "salman"
b = "sakf454"
print(a.isalpha())
print(b.isalpha())
```

### **18. isdigit()**
Returns True if all characters are digits.

```python
a = "9545845"
b = "ksf959458"
print(a.isdigit())
print(b.isdigit())
```

### **19. isalnum()**
Returns True if all characters are alphabets or digits.

```python
a = "saldfj4566"
b = "sdkfjf-"
print(a.isalnum())
print(b.isalnum())
```

### **20. capitalize()**

Capitalizes only the first character.

```python
a = "muhammad salman"
a.capitalize()
```

### **21. title()**
Capitalizes the first letter of every word.

```python
a = "muhammad salman"
a.title()
```

### **22. swapcase()**
Converts uppercase letters to lowercase and lowercase letters to uppercase.

```python
a = "Muhammad Salman"
a.swapcase()
```

### **23. format()**
Formats a string by inserting values into placeholders.

```python
a = "my name is {} and age is {}."
print(a.format("salman", 19))
a = "my name is {1} and age is {0}"
print(a.format(19, "salman"))
```

### **24. index()**
Returns the index of the first occurrence of a substring.

Unlike find(), it raises an error if the substring is not found.

```python
a = "salman"
print(a.index("a"))
print(a.index("k"))
```

# **Quick Revision**

| Function/Method | Purpose               | Returns                                  |
| --------------- | --------------------- | ---------------------------------------- |
| `len()`         | Counts characters     | `int`                                    |
| `upper()`       | Converts to uppercase | `str`                                    |
| `lower()`       | Converts to lowercase | `str`                                    |
| `split()`       | Splits a string       | `list`                                   |
| `join()`        | Joins an iterable     | `str`                                    |
| `find()`        | Finds a substring     | `int` (`-1` if not found)                |
| `index()`       | Finds a substring     | `int` (raises `ValueError` if not found) |
| `replace()`     | Replaces text         | `str`                                    |
| `count()`       | Counts occurrences    | `int`                                    |
| `startswith()`  | Checks prefix         | `bool`                                   |
| `endswith()`    | Checks suffix         | `bool`                                   |

### 6 - Write a program to count the number of words in a string without split()

```python
a = input("enter sentence to count its word")
tem = ""
words = []

for i in a:
    if i != " ":
        tem = tem+i 
    else:
        words.append(tem)
        tem = ""
words.append(tem)
        
print(words)
print(len(words))
```

### 7 - Write a python program to convert a string to title case without using the title()

```python
# Method - 1
a = input("enter any sentence")

tit = ""
temp = ""
for i in a:
    if i != " ":
        temp+=i
    else:
        tit = tit + " " + temp.capitalize()
        temp = "" 

tit = tit + " " + temp.capitalize()
   
print(tit)
```

```python
# Method - 2

a = input("Enter sentence")
tit = ""
for i in a.split():
    result = i[0].upper()+i[1:]
    tit = tit + " " + result

print(tit)    
```
