# **Lists**
List is a data type where you can store multiple items under 1 name. LIsts are mutable as we can modify the list and can make changes with it.<br>

# **Creating List**
List can simply created by storing multiples data types in a square brackets.

```python
li = [12, 5, 6, 8]
print(type(li))
li = [12, 5, None, True, "salman"]
print(type(li))
print(li)
```

# **Types of List**

### **Types of list by dimension**
There are three types of list on the basis of dimension of the list.

### **1. 1D List**

```python
a = [1, 2, 3, 4, 5]
print(a)
print(type(a))
```

### **2. 2D List**

```python
a = [[2, 3], [1, 3], [2, 5], [5, 6]]
print(a)
print(type(a))
```

### **3. 3D List**

```python
a = [[[2, 4], [3, 4]], [[2, 4], [4, 7]]]
print(a)
print(type(a))
```

### **Types of list by datatype**

### **1. Homogenouse**
A list which have same data type items.

```python
a = ["salman", "naveed", "junaid", "ubaid", "nadeem"]
print(type(a))
```

### **2. Heterogenous**
A list which have different data types items. 

```python
a = ["salman", 3, None, True]
print(type(a))
```

# **List Indexing**

### **1. Positive Indexing**

```python
a = [2, 3, 5, 6] # 1D
print(a[0])

b = [[2, 3], [3, 4]]   # 2D
print(b[1][0])


c = [[[2, 3], [2, 5]], [[2, 4], [3, 4]]]   # 3D
print(c[1][0][1])
```

### **2. Negative Indexing**

```python
a = [3, 4, 5, 6, 6]

print(a[-1])
print(a[-4])
```

# **List Slicing**

```python
a = [3, 5, 7, 4, 8, 5, 7, 8, 3]
print(a[0:4])

print(a[2:6])

print(a[1:9:2])
```

```python
a = [3, 4, 5, 6, 7, 8, 9]  # Reverse a list
print(a[::-1])
```

# **Adding items to List**

### **1. append()** 
It is used to add a single item at the end of list

```python
a = [1, 2, 3, 4, 5, 6]
a.append(7)   # as lists are mutable so it modify the original list
print(a)
```

### **2. extend()**
It is used to add items(iterable) at the end of list

```python
a = [3, 4, 5, 6]
a.extend([8, 9])
print(a)
```

### **3. insert()**
It is used to add item at specific index.<br>
list.insert(index, item)

```python
a = [1, 2, 3, 4, 5, 6]
a.insert(1, "salman")
print(a)
```

# **Editing items in List**

### **1. Editing Single Item**

```python
a = ["slaman", "naveed", "junaid", 3, 5, 6]

a[0] = "muhammad"

a[4] = "Salman"

print(a)
```

```python
a = ["muahammad", "salman", "junaid", "shamir", "nadeem", "waseem"]

a[1:4] = [3, 5, 7]

print(a)
```

# **Deleting items from List**

### **1. Deleting Single Item**

```python
a = [3, 4, 5, 6, 7, 8, 6]

del a[1]

print(a)
```

### **2. Deleting Multiple Item**

<b><h1>List Comprehension</h1></b>
List Comprehension is a concise and efficient way to create a new list by writing the logic in a single line instead of using multiple lines of code with loops.

image.png


```python
a = [i for i in range(1,6)]
print(a)
```

```python
a = [i for i in range(1, 11) if i%2==0 ]
print(a)
```

```python
a = ["even" if i % 2 == 0 else "odd" for i in range(1, 11)]

print(a)
```

```python
numbers = [1,2,3,4,5]

result = ["Even" if num % 2 == 0 else "Odd" for num in numbers]

print(result)
```

```python
a = "Python"

letters = [i.upper() for i in a]

print(letters)
```

# **List Functions and Methods**

### **1. len()**
The len() function returns the total number of elements in a list.

```python
a = [2, 3, 4, 5, 6, 6]
print(len(a))
```

### **2. min()**
Returns the smallest element in a list.

```python
a = [1, 2, 4, 5, 6, 7, 8, 9]
print(min(a))
```

### **3. max()**
Returns the largest element in a list.

```python
a = [2, 3, 4, 5, 6, 7, 8]
print(max(a))
```

### **4. sorted()**
Returns a new sorted list without modifying the original list to sort the element in ascending or descending order.

```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]
print(sorted(a, reverse=True))     # reverse true is used for descending order

print(sorted(a, reverse=False))   # reverse false is used for ascending order
```

### **5. count()**
Returns the number of times a specified element appears in the list.

```python
a = [3, 4, 5, 6, 7, 8, 9, 4]
print(a.count(4)) 
```

### **6. index()**
Returns the index of the first occurrence of an element.<br>
**Syntax**<br>
list.index(value, start, end)

| Parameter | Description              |
| --------- | ------------------------ |
| `value`   | Element to search.       |
| `start`   | Optional starting index. |
| `end`     | Optional ending index.   |


```python
a = [1, 2, 3, 4, 5, 2, 6, 7, 8, 9, 1]
print(a.index(2))
print(a.index(1, 4, ))
print(a.index(2, 2, 8))
```

### **7. reverse()**
Reverses the order of elements in the original list.

```python
a = [1, 2, 3, 4, 5, 2, 6, 7, 8, 9]
a.reverse()
print(a)
```

### **8. copy()**
Creates a shallow copy of a list.

```python
a = [1, 2, 3, 4, 5, 2, 6, 7, 8, 9]
b = a.copy()
print(b)
```

### **9. zip()**
The zip() function combines elements from two or more iterables into pairs (tuples).<br>
zip(iterable1, iterable2, ...)

```python
a = [1, 2, 3, 4, 5]
b = [6, 7, 8, 9, 0]
c = zip(a, b)
print(list(c))
c = zip(a, b)
print(dict(c))
```

```python
a = [1, 2, 3, 4, 5]
b = [6, 7, 8, 9, 0]
c = zip(a, b)

for i,j in c:
    print(i,j)
```

### **10. pop()**
Removes and returns an element at a specified index.

```python
a = ["salman", "junaid", "ubaid"]
a.pop(0)
```

```python
print(a)
```

### **11. remove()**
Removes the first occurrence of a specified value.

```python
a = ["salman", "ubaid", "naveed", "junaid"]
a.remove("salman")
print(a)
```

### **12. clear()**
Removes all elements from a list

```python
a = ["salman", "ubaid", "naveed", "junaid"]
a.clear()
print(a)
```

### **13. sum()**
Sum all the numbers in a list.

```python
a = [1, 2, 3, 4]
print(sum(a))
```

### **14. sort()**
Sort the original list.

```python
a = [1, 2, 3, 4, 5,6, 7]
a.sort()
print(a)
```
