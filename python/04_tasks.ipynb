### `Problem 1:` Combine two lists index-wise(columns wise)

Write a program to add two lists index-wise. Create a new list that contains the 0th index item from both the list, then the 1st index item, and so on till the last element. any leftover items will get added at the end of the new list.

`Given List:`
```
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]
```

`Output:`
```
[['M','y'], ['na', me'], ['i', 's'], ['Kh', 'an']]
```

````python
# Write code here
# method - 1
list1 = ["M", "na", "i", "Kh"]
list2 = ["y", "me", "s", "an"]

result = []

for i in range(0, len(list1)):
    result.append([list1[i], list2[i]])
    
    
print(result)


# method - 2
[[list1[i], list2[i]] for i in range(0, len(list1))]
````

### `Problem 5:` You are given a list of integers. You are asked to make a list by running through elements of the list by adding all elements greater and itself.

i.e. Say given list is `[2,4,6,10,1]`
resultant list will be `[22,20,10,23]`.

For 1st element `2` ->> these are greater `(4+6+10)` values and `2` itself so on adding becomes `22`.

For 2nd element `4` ->> greater elements are `(6, 10)` and `4` itself, so on adding `20`

like wise for all other elememts.

`[2,4,6,10,1]`-->`[22,20,16,10,23]`




````python
# Write code here

a = [2,4,6,10,1]

result = []

for i in range(0, len(a)):
    sum = a[i]
    for j in range(i, len(a)):
        if a[i]<a[j]:
            sum = sum+a[j]
    result.append(sum)
                  
print(result)

````

### `Problem 6:` Find list of common unique items from two list. and show in increasing order

`Input`

```
num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]
```

`Output:`
```
[34, 67, 89]
```


````python
# Write code here

num1 = [23,45,67,78,89,34]
num2 = [34,89,55,56,39,67]

result = []

for i in range(0, len(num1)):
    for j in range(0, len(num2)):
        if num1[i]==num2[j]:
            result.append(num2[j])

result.sort()    
print(result)
````

### `Problem 7:` Sort a list of alphanumeric strings based on product value of numeric character in it. If in any string there is no numeric character take it's product value as 1.

`Input:`
```
['1ac21', '23fg', '456', '098d','1','kls']
```

`Output:`
```
['456', '23fg', '1ac21', '1', 'kls', '098d']

```

````python
# Write code here

l = ['1ac21', '23fg', '456', '098d','1','kls']


n = "0123456789"

lis = []

for i in l:
    pro = 1
    for j in i:
        for k in n:
            if j==k:
                pro=pro*int(j)
                
    lis.append(pro)
                
           
        
        
print(lis)
print(l)




ind = []
    

for i in range(0, len(l)):
    ind.insert(lis[i], l[i])
    
    
ind.reverse()

print(ind)




````

### `Problem 10:` Add Space between Potential Words.

**Example:**

Input:

['campusxIs', 'bestFor', 'dataScientist']

```bash
['campusx Is', 'best For', 'data Scientist']
```

````python
# Code here

a = ['campusxIs', 'bestFor', 'dataScientist']

li = []


for i in a:
    word = ''
    for j in i:
        if j in "ABCDEFGHIJKLMNOPQRSTUVWXZYZ":
            word = word+" "+j
        else:
            word = word + j
    li.append(word)


print(li)
````

### `Problem 11:` Write a program that can perform union operation on 2 lists

**Example:**

Input:

```bash
[1,2,3,4,5,1]
[2,3,5,7,8]
```

Output:
```bash
[1,2,3,4,5,7,8]
```

````python
# Write code here

# method - 1

a = [1,2,3,4,5,1]
b = [2,3,5,7,8]

union = []

for i in a:
    status = True
    for j in union:
        if i == j:
            status=False
    if status == True:
        union.append(i)
        
for i in b:
    status = True
    for j in union:
        if i == j:
            status = False
    if status == True:
        union.append(i)
        
union.sort()
print(union)
````


### `Problem 12:` Write a program that can find the max number of each row of a matrix

**Example:**

Input:

```bash
[[1,2,3],[4,5,6],[7,8,9]]
```

Output:
```bash
[3,6,9]
```

````python
# Write code here
from functools import reduce

l = [[1,2,3],[4,5,6],[7,8,9]]

max = []

for i in l:
    num = reduce(lambda a,b: a if a>b else b, i)
    max.append(num)
    
    
print(max)
````

### `Problem 14:` Write a list comprehension that can transpose a given matrix

matrix = [<br>

          [4,5,6],<br>
          [7,8,9]<br>]<br>

[1, 4, 7]<br>
[2, 5, 8]<br>
[3, 6, 9]<br>

````python
# Write code here

matrix = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
]


[
    [matrix[j][i] for j in range(0, 3)]
    for i in range(0,3)
]
    
````

### `Problem 15:` Write a list comprehension that can flatten a nested list

Input<br>
matrix = [<br>
  [1,2,3],<br>
          [4,5,6],<br>
          [7,8,9]<br>]<br>

Output:<br>
[1, 2, 3, 4, 5, 6, 7, 8, 9]

````python
# Write code here

matrix = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
]


[
    j for i in matrix
    for j in i
]
````
