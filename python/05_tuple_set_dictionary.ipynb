<b><h1><center>Tuples in Python</center></h1></b>

# **Tuple**
A tuple in Python is similar to a list. The difference between the two is that we cannot change the elements of a tuple once it is assigned whereas we can change the elements of a list.

In short, a tuple is an immutable list. A tuple can not be changed in any way once it is created.
<br>

# **Creating Tuple**
Tuple can simply created by storing multiples data types in a parenthesis.

```python
a = (1, 2, 3, 4)
b = ("hello", "world")
print(type(a))
print(type(b))
print(b)
print(a)
```

```python
# creating single element tuple
a = ("hello",)  # we have to use comma for tuple in a single element
print(type(a))
b = (2, )
print(type(b))
c = ("salman")    # its a string 
print(type(c))
```

# **Types of Tuple**

### **Types of tuple by dimension**
There are three types of tuple on the basis of dimension of the tuple.

### **1. 1D List**

```python
a = (1, 2, 3, 4)
print(a)
```

### **2. 2D List**

```python
a = ((1, 3),(2, 3))
print(a)
```

### **3. 3D List**

```python
a = (((1, 2),(1, 2), (1, 5), (6, 7)))
print(a)
```

### **Types of tuple by datatype**

### **1. Homogenouse**
A tuple which have same data type items.

```python
a = ("salman", "naveed", "junaid", "ubaid", "nadeem")
print(type(a))
print(a)
```

### **2. Heterogenous**
A tuple which have different data types items. 

```python
a = ("salman", 3, None, True)
print(type(a))
print(a)
```

# **Tuple Indexing**

### **1. Positive Indexing**

```python
a = (3, 5, 7, 8)
print(a[0])
print(a[1])
print(a[3])
```

### **2. Negative Indexing**

```python
a = (3, 5, 7, 8)
print(a[-1])
print(a[-2])
print(a[-4])
```

# **Tuple Slicing**

```python
a = (3, 4, 6, 7, 9, 7)
print(a[0:4])
print(a[3:])
```

# **Adding items to Tuple**
We cannot add items to the tuple because tuples are immutable as we cannot make changes with it.

```python
a = (1, 2, 3, 4)
a.append(3)
```

# **Editing items in Tuple**
We cannot edit any items in a tuple due to imutability of the tuple.

```python
a = (1, 3, 4, 5, 6)
a[1] = 0
```

# **Deleting items from Tuple**
We also cannot delete any item from the tuple this is all due to imutability of the tuple.

```python
a = (2, 3, 4, 5)
del a[2]
```

# **Deleting Tuple**
We can delete the whole tuple.

```python
a = (2,3,4 ,5 ,6)
print(a)

del a    # Tuple deleted
print(a)
```

# **Tuple unpacking**
It used to asign variable to each value.

```python
a, b, c = (1, 3, 5)
print(a)
print(b)
print(c)
```

```python
a, *other = (2, 3, 4, 5, 6, 7, 9)
print(a)
print(other)
```

# **Operation on Tuple** 

### **1. Concatenation (+)**
It join the two list into one tuple

```python
a = (1, 2, 3, 5)
b = (4, 6, 8, 9)
 
a+b
```

### **2. Repetition (*)**

```python
a = (1, 2, 3)
a*3
```

### **3. Membership**

```python
a = ("salman", "naveed", "junaid")
"salman" in a
```

### **4. Iteration**

```python
a = ("salman", "naveed", "junaid", "nadeem", "waseem", "farhad", "ubaid")
for i in a:
    print(i)
```

# **Tuple Functions and Methods**

### **1. len()**
The len() function returns the total number of elements in a tuple.

```python
a = (2, 4, 5)
len(a)
```

### **2. min()**
Returns the smallest element in a tuple.

```python
a = (2, 4, 5)
min(a)
```

### **3. max()**
Returns the largest element in a tuple.

```python
a = (2, 4, 5)
max(a)
```

### **4. sorted()**
Returns a new sorted tuple without modifying the original tuple to sort the element in ascending or descending order.

```python
a = (2, 4, 5, 9, 1, 3)
sorted(a)
```

### **5. count()**
Returns the number of times a specified element appears in the tuple.

```python
a = (1, 2, 3, 3, 4, 3)
a.count(3)
```

### **6. index()**
Returns the index of the first occurrence of an element.<br>
**Syntax**<br>
tuple.index(value, start, end)

| Parameter | Description              |
| --------- | ------------------------ |
| `value`   | Element to search.       |
| `start`   | Optional starting index. |
| `end`     | Optional ending index.   |


```python
a = (1, 2, 3, 4, 5)
a.index(4)
```

### **9. zip()**
The zip() function combines elements from two or more iterables into pairs (tuples).<br>
zip(iterable1, iterable2, ...)

```python
a = (2, 3, 4, 5)
b = (3, 4, 5, 6)

c = zip(a, b)
tuple(c)
```

```python
a = (2, 3, 4, 5)
b = (3, 4, 5, 6)

for i, j in zip(a, b):
    print(i, j)
```

### **13. sum()**
Sum all the numbers in a list.

```python
a = [1, 2, 3, 4]
print(sum(a))
```

# **Set**

A set is an unordered collection of items. Every set element is unique (no duplicates) and must be immutable (cannot be changed).

However, a set itself is mutable. We can add or remove items from it.

Sets can also be used to perform mathematical set operations like union, intersection, symmetric difference, etc.
<br>

# **Creating Set**
Set can simply created by storing multiples data types in a curly brackets.

```python
a = {1, 2, 4, 5}
print(a)
print(type(a))
```

```python
# Set ignore the duplicates
a = {1, 2, 2, 2, 3, 4}
print(a)
```

# **Types of Set**

### **1. Homogenouse**
A set which have same data type items.

```python
a = {"salman","farhad", "nadeem"}
print(type(a))
```

### **2. Heterogenous**
A set which have different data types items. 

```python
a = {"salman", 3, None, True}
print(type(a))
print(a)

# Set can`t have mutabile datatypes
a = {2, 4, 5, [3, 5]}    # error because list is mutable and set cannot have mutable items

a = {1, 34, {2, 5}}   # error because the inside set is also mutable
```

# **Set Indexing**
We cannot do indixing with set because set is the collection of unordered items and the items have not specific index.

```python
a = {1, 2, 3, 4}
a[3]
```

# **Set Slicing**
We cannot do slicing with set because set is the collection of unordered items and the items have not specific index.

```python
a = {3, 4, 5, 6, 7}
a[3:5]
```

# **Adding items to Set**

```python
a = {1, 2, 4, 5, 6}
a.add(3)
print(a)
```

# **Editing items in Set**
We cannot edit item from set due to unordered property of set.

```python
a = {2, 3, 4, 5}
a[1] = 3
```

# **Deleting items from Set**
We cannot delete item from set by del function but there are some methods of set through which we can delete item.

```python
a = {3, 4, 5, 6, 7}
del a[0]
```

```python
a = {4, 5, 6, 7, 8, 9}
a.discard(5)
print(a)
```

# **Deleting Set**

```python
a = {1, 2, 3}
del a
print(a)   # set deleted
```

# **Set unpacking**
It used to asign variable to each value.

```python
a, b, c = {1, 2, 3}
print(a)
print(b)
print(c)
```

```python
a,*other = {2, 3, 4}
print(a)
print(other)
```

### **2. Intersection(&)**

Return all elements from both sets except common elements.


```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a ^ b
```

### **4. Membership**

```python
a = {1, 2, 3, 4}
1 in a
```

# **Set Functions and Methods**

### **1. len()**
The len() function returns the total number of elements in a set.

```python
a = {1, 2, 3, 4, 5, 6}
len(a)
```

### **2. min()**
Returns the smallest element in a set.

```python
a = {1, 2, 3, 4, 5, 6}

min(a)
```

### **3. max()**
Returns the largest element in a set.

```python
a = {1, 2, 3, 4, 5, 6}

max(a)
```

### **4. sorted()**
Returns a new sorted tuple without modifying the original tuple to sort the element in ascending or descending set.

```python
a = {1, 2, 3, 4, 5, 6}
sorted(a)
```

### **5. sum()**
Sum all the numbers in a set.

```python
a = {1, 2, 3, 4, 5, 6}
sum(a)
```

### **6. union()**
The union() method returns a new set containing all unique elements from two or more sets.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.union(b)
```

### **7. update()**
Update the original first set by adding the other set. 

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.update(b)
print(a)
```

### **8. intersection**
Returns a new set containing only the common elements between two or more sets.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.intersection(b)
```

### **9. intersection_update()**
Updates the original set by keeping only the common elements.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.intersection_update(b)
print(a)
```

### **10. differenc()**
Returns a new set containing elements that are present in the first set but not in the second.


```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.difference(b)
```

### **11. difference_update()**
Removes common elements from the original set.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.difference_update(b)
print(a)
```

### **12. symetric_difference()**
Returns elements that exist in either set but not in both.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.symmetric_difference(b)
```

### **13. symetric_differenc_update()**
Updates the original set by keeping only non-common elements.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.symmetric_difference_update(b)
print(a)
```

### **14. isdisjoint()**
Checks whether two sets have no common elements.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4}
a.isdisjoint(b)
```

### **15. issubset()**
Checks whether all elements of one set are present in another set.

```python
a = {1, 2, 3, 4, 5}
b = {6, 7, 8, 9, 0, 4, 1, 2, 3, 4, 5}
a.issubset(b)
```

### **16. issuperset()**
Checks whether one set contains all elements of another set.

```python
a = {1, 2, 3, 4, 5, 6, 7, 8, 9, 0, 4}
b = {6, 7, 8, 9, 0, 4}

a.issuperset(b)
```

### **17. copy()**
Creates a shallow copy of a set.

```python
a = {3, 4, 5, 6}
b = a.copy()
print(a)
```

### **18. discard()**
Removes an element if it exists.

```python
b = {6, 7, 8, 9, 0, 4, }
b.discard(0)
print(b)
```

### **19. remove()**
Removes a specified element.

If the element is not present, it raises a KeyError.

```python
b = {6, 7, 8, 9, 0, 4, }

b.remove(6)
print(b)
```

### **20. pop()**
Removes and returns a random element from the set.

```python
a = {3, 4, 5, 6, 7, 8, 9}
a.pop()
```

### **21. clear()**
Removes all elements from the set.

```python
a = {3, 4, 5, 6, 7, 8, 9}
a.clear()
print(a)
```

# **Quick Revision**

| Method                          | Purpose                                      | Modifies Original Set? | Returns         |
| ------------------------------- | -------------------------------------------- | ---------------------- | --------------- |
| `union()`                       | Combine unique elements                      | ❌ No                   | New set         |
| `intersection()`                | Common elements                              | ❌ No                   | New set         |
| `intersection_update()`         | Keep common elements                         | ✅ Yes                  | `None`          |
| `difference()`                  | Elements only in first set                   | ❌ No                   | New set         |
| `difference_update()`           | Remove common elements                       | ✅ Yes                  | `None`          |
| `symmetric_difference()`        | Non-common elements                          | ❌ No                   | New set         |
| `symmetric_difference_update()` | Keep non-common elements                     | ✅ Yes                  | `None`          |
| `isdisjoint()`                  | No common elements?                          | ❌ No                   | `bool`          |
| `issubset()`                    | Is this set contained in another?            | ❌ No                   | `bool`          |
| `issuperset()`                  | Does this set contain another?               | ❌ No                   | `bool`          |
| `copy()`                        | Create a shallow copy                        | ❌ No                   | New set         |
| `discard()`                     | Remove element (no error if absent)          | ✅ Yes                  | `None`          |
| `remove()`                      | Remove element (raises `KeyError` if absent) | ✅ Yes                  | `None`          |
| `pop()`                         | Remove and return a random element           | ✅ Yes                  | Removed element |
| `clear()`                       | Remove all elements                          | ✅ Yes                  | `None`          |

<b><h1><center>Dictionary in Python</center></h1></b>
Dictionary in Python is a collection of keys values, used to store data values like a map, which, unlike other data types which hold only a single value as an element.

In some languages it is known as map or assosiative arrays.

dict = { 'name' : 'nitish' , 'age' : 33 , 'gender' : 'male' }<br>

# **Creating Dictionary**


```python
a = {"name": "salman", "age":19, "marks":90}
print(a)
print(type(a))
```

# **Types of Dictionary**

### **Types of set by dimension**
There are two types of dictionary on the basis of dimension of the Dictionary.

### **1. 1D Dictionary**

```python
a = {"name": "salman", "age": 56}
print(a)
```

### **2. 2D Dictionary**

```python
a = {"name": "salman", "age":19, "marks":{"math":90, "physics":56, "computer_science":89}}
print(a)
print(type(a))
```

# **Accessing values**
We can access the values of any key by providing key of that value.

```python
# for 1D
a = {"name": "salman", "age": 56}

print(a["name"])
print(a["age"])
```

```python
# for 2D

a = {"name": "salman", "age":19, "marks":{"math":90, "physics":56, "computer_science":89}}

print(a["name"])
print(a["marks"]["math"]) 

```

```python
a = {"name": "salman", "age":19, "marks":{"math":90, "physics":56, "computer_science":89}}
a["salary"]   # it throw an error when the key does not exsist in dictionary
```

```python
a = {"name": "salman", "age":19, "marks":{"math":90, "physics":56, "computer_science":89}}
a.get("salary")  # it is also used to access the value of any key but it does not throw error when the key not exsist.
a.get("name")
```

# **Adding key-value pair**
WE can add any key-value pair to the dictionary.

```python
a = {"name": "salman", "age": 56}

a["marks"]=90

print(a)
```

# **Editing key-value pair**
We can edit any value of key.

```python
a = {"name": "salman", "age": 56}

a["name"]="naveed"

print(a)
```

# **Deleting key-value pair**
We can delete any key-value pair from a given dictionary.

```python
a = {"name": "salman", "age": 56}

del a["name"]


print(a)
```

# **Deleting dictionary**
We can delete the whole dictionary.

```python
a = {"name": "salman", "age": 56}

del a  # dictionary deleted

print(a)
```

# **Operation on Dictionary** 

### **1. Membership**
Membership operation work on the basis of key of the dictionary.

```python
a = {"name": "salman", "age": 56}

print("name" in a)

print("age" in a)
```

### **2. Iteration**
We can also iterate through the keys of the dictionary.

```python
a = {"name": "salman", "age": 56}

for i in a:  
    print(i)
```

```python
    
for i in a:
    print(i, a[i])
```

# **Dictionary Functions and Methods**

```python
# pop/popitem/clear/len/sorted/items/keys/values/update/zip
```

### **1. pop()**
The pop() method removes a specified key from a dictionary and returns its value.

```python
a = {"name": "salman", "age": 56, "marks":90}
a.pop("marks")
print(a)
```

### **2. popitem()**
Removes and returns the last inserted key-value pair.

```python
a = {"name": "salman", "age": 56}

a.popitem()
print(a)
```

### **3. clear()**
Removes all key-value pairs from a dictionary.

```python
a = {"name": "salman", "age": 56, "marks":90}

a.clear()
print(a)
```

### **4. len()**
Returns the total number of key-value pairs.

```python
a = {"name": "salman", "age": 56, "marks":90}
print(len(a))
```

### **5. sorted()**
Returns a sorted list of dictionary keys.

```python
a = {"name": "salman", "age": 56, "marks":90}
sorted(a)
```

### **6. items()**
Returns all key-value pairs as dictionary items.We can easily access any key value.

```python
a = {"name": "salman", "age": 56, "marks":90}
c = a.items()
print(c)
```

```python
for i,j in c:
    print(i,j)
```

### **7. keys()**
Returns all keys of a dictionary.

```python
a = {"name": "salman", "age": 56, "marks":90}
a.keys()
```

### **8. values()**
Returns all values of a dictionary.

```python
a = {"name": "salman", "age": 56, "marks":90}
a.values()
```

### **9. update()**
Adds new key-value pairs or updates existing ones.

```python
a = {"name": "salman", "age": 56}
b = {"marks":90}
a.update(b)
print(a)
```

```python
a = {"name": "salman", "age": 56}
b = {"marks":90, "district":"kpk"}
a.update(b)
print(a)
```

### **10. zip()**
Combines multiple iterables element-by-element.

```python
keys = ["name", "age", "city"]
values = ["Salman", 18, "DIK"]

c = dict(zip(keys, values))

print(c)
```
