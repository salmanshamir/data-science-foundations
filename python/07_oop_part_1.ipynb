# **Magic Methods**
Magic functions are special methods in Python that have double underscores (__) before and after their names.

They are automatically called by Python when certain operations are performed on an object.

**In simple words:**

Magic methods allow Python objects to respond to built-in operations like creating objects, printing, comparing, adding, calculating length, and much more.

They are called magic because you usually don't call them directly.

They are also called Dunder methods as "Dunder" means Double UNDERscore.

| ---------------- | -------------------------------- |
| `__init__()`     | Initialize an object             |
| `__str__()`      | String for `print()`             |

| `__len__()`      | Support `len()`                  |
| `__add__()`      | `+` operator                     |
| `__sub__()`      | `-` operator                     |
| `__mul__()`      | `*` operator                     |
| `__truediv__()`  | `/` operator                     |
| `__eq__()`       | `==` comparison                  |
| `__lt__()`       | `<` comparison                   |
| `__gt__()`       | `>` comparison                   |
| `__contains__()` | `in` operator                    |
| `__getitem__()`  | Access with `[]`                 |
| `__setitem__()`  | Assign with `[]`                 |
| `__call__()`     | Make an object callable          |
| `__del__()`      | Called before object destruction |

### **2. __str__()**
Returns a human-readable string representation of an object.

```python

class Demo:
    def __str__(self):
        return "print executed"
        
obj = Demo()

print(obj)
```

### **3. __repr__()**
Returns the official string representation of an object.

```python

class Demo:
    def __repr__(self):
        return "repr"
        
obj = Demo()
print(repr(obj))
```

### **4. __len__()**
Defines what happens when len() is called on an object.

```python

class Demo:
    def __len__(self):
        return 33
        
obj = Demo()
print(len(obj))
```

### **5. __add__()**
Defines the behavior of the + operator for objects.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __add__(self, other):
        return self.num + other.num
        
obj1 = Demo(4)
obj2 = Demo(7)

print(obj1+obj2)

```

### **6. __sub__()**

Defines the behavior of the - operator

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __sub__(self, other):
        return self.num - other.num
        
obj1 = Demo(4)
obj2 = Demo(7)

print(obj1-obj2)

```

### **7. __mul__()**

Defines the behavior of the * operator.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __mul__(self, other):
        return self.num * other.num
        
obj1 = Demo(4)
obj2 = Demo(7)

print(obj1*obj2)

```

### **8. __truediv__()**

Defines the behavior of the / operator.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __truediv__(self, other):
        return self.num / other.num
        
obj1 = Demo(4)
obj2 = Demo(7)

print(obj1/obj2)

```

### **9. __eq__()**

Defines the behavior of the == operator.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __eq__(self, other):
        return self.num == other.num
        
obj1 = Demo(4)
obj2 = Demo(4)

print(obj1==obj2)

```

### **10. __lt__()**

Defines the behavior of the < operator.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __lt__(self, other):
        return self.num < other.num
        
obj1 = Demo(6)
obj2 = Demo(4)

print(obj1<obj2)

```

### **11. __gt__()**

Defines the behavior of the > operator.

```python

class Demo:
    def __init__(self, num):
        self.num = num
    
    def __gt__(self, other):
        return self.num > other.num
        
obj1 = Demo(41)
obj2 = Demo(4)

print(obj1>obj2)

```

### **12. __contains__()**

Defines the behavior of the in operator.

```python

class Demo:
    def __init__(self, name):
        self.name = name
    
    def __contains__(self, letter):
        return letter in self.name
        
obj1 = Demo("salman")


print("s" in obj1)
print("p" in obj1)

```

### **13. __getitem__()**

Defines what happens when using square brackets ([]) to retrieve an item

```python
class Demo:
    def __init__(self, name):
        self.name = name
    
    def __getitem__(self, letter):
        return self.name[letter]
        
obj1 = Demo("salman")


print(obj1[1])
print(obj1[0])
print(obj1[4])

```

### **14**. __setitem__()

Defines what happens when assigning a value using square brackets.

```python
class Demo:
    def __init__(self):
        self.dic = {}
    
    def __setitem__(self, key, value):
        self.dic[key] = value
        
obj1 = Demo()

obj1["Name"]="Salman"

obj1.dic


```

### **15. __call__()**

Makes an object callable like a function.

```python
class Demo:
    def __call__(self, letter):
        return "call execute hogaya", letter
        
obj1 = Demo()

obj1("l")


```

### **16. __del__()**

Called when an object is about to be destroyed (garbage collected). It is generally not recommended to rely on it for important cleanup because the timing of object destruction can vary.

```python
class Demo:
    def __del__(self):
        print("no dont try")
        
obj1 = Demo()

del obj1

```

### 2 - Create a fraction data type which can add sub multiply and divide fraction.

```python
class Fraction:
    def __init__(self, x, y):
        self.num = x
        self.den = y
        
    def __str__(self):
        return f"{self.num}/{self.den}"
    
    def __add__(self, other):
        num = self.num*other.den+other.num*self.den
        den = self.den*other.den
        return f"{num}/{den}"
    
    
    def __sub__(self, other):
        num = self.num*other.den-other.num*self.den
        den = self.den*other.den
        return f"{num}/{den}"
    
    def __mul__(self, other):
        num = self.num*other.num
        den = self.den*other.den
        return f"{num}/{den}"
    
    def __truediv__(self, other):
        num = self.num*other.den
        den = self.den *other.num
        return f"{num}/{den}"
    
    def convert_to_decimal(self):
        return self.num/self.den
        
        
  
```

```python
fr1 = Fraction(2, 3)
fr2 = Fraction(3, 4)
```

```python
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
```
