# **Private variable**
A private variable is an attribute that is intended to be accessed only inside the class and cannot be access easily outside the class. It helps hide the internal data of an object from direct access by external code.

In Python, a private variable is created by adding two underscores (_ _) before the variable name.And when we want to access it inside the class then we use double underscore (__) before the variable name.

```python
class Demo:
    def __init__(self):
        self.name = "salman"
        self.__pin = 123
        
    def show(self):
        print(self.name)
        print(self.__pin)
        
        
obj = Demo()
obj.name       # we can access the name which is not private and it is called public attribure
```

```python
obj.pin        # we cannot access the pin because it it private and it give us an error that demo has no attribute pin
```

### **Can We Still Access It?**
Yes

When Python sees:

**self.__pin**

it internally changes it to something like:

**self._Demo__pin**

As here Demo is class name and pin is attribute so python changes every private variable to this formate.
_classname__attributename

####  **So we can access it outside the class by using _classname__attributename**




```python
class Demo:
    def __init__(self):
        self.name = "salman"
        self.__pin = 123
        
    def show(self):
        print(self.name)
        print(self.__pin)
        
        
obj = Demo()
obj._Demo__pin
```


### **Why Does Python Do This?**

Python follows the philosophy:

**"We are all consenting adults here."**

Instead of completely blocking access, Python warns:

"This is an internal attribute. Please don't access it directly."

# **Getter and Setter methods** 
Get and Set are methods through which we can access the value and modify the private attribute.

### **Get Method**
A Getter is a method used to access (read) the value of a private variable.


```python
class Demo:
    def __init__(self):
        self.name = "salman"
        self.__pin = 123
        
    def get_pin(self):
        print(self.__pin)
        
        
obj = Demo()
obj.get_pin()
```

### **Set Method**
A Setter is a method used to modify (update) the value of a private variable, usually after validating the new value.

```python
class Demo:
    def __init__(self):
        self.name = "salman"
        self.__pin = 123
        
    def get_pin(self):
        print(self.__pin)
        
    def set_pin(self, new_pin):
        self.__pin = new_pin
        
obj = Demo()
obj.get_pin()
obj.set_pin(235)
obj.get_pin()
```

# **Encapsulation in OOP**
Encapsulation is the process of wrapping (combining) data (attributes) and the methods (functions) that operate on that data into a single unit called a class, while restricting direct access to some of the object's data.

**Is Encapsulation Only About Private Variables?**

No.

Many beginners think:

**Encapsulation = Private variables**
This is not completely correct.

Encapsulation has two parts:

Combining data and methods into one class.

Restricting direct access when necessary (often using private variables).

**Even a class with only public attributes still demonstrates the first part of encapsulation because it groups related data and behavior together.**

Private variables simply make encapsulation stronger by protecting the data.

```python
class Demo:
    def __init__(self):
        self.name = "salman"
        self.__pin = 123
        
    def get_pin(self):
        print(self.__pin)
        
    def set_pin(self, new_pin):
        self.__pin = new_pin
        print("you successfully change the pin")
        
    def get_name(self):
        print(self.name)
        
        
   
obj = Demo()
obj.get_pin()
obj.set_pin(235)
obj.get_name()
```

# **Static Variable Vs Instant Variable**
A variable inside a class is called an attribute.

There are two main types:

- **Instance Variables**

- **Class Variables (Static Variables)**

The biggest difference is where they are stored and who owns them.

### **Instant Variable**
An instance variable is a variable that belongs to an object (instance).

Every object gets its own separate copy.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
```

The keyword `self` tells Python:

"Store this variable inside this particular object."

`name` and `age` are different for every object.

```python
obj1 = Student("salman", 19)
obj2 = Student("Naveed", 20)
print(obj1.name)
print(obj2.name)
```

### **Class Variable (Static Variable)**
A class variable belongs to the class itself, not to any particular object.

Only one copy exists, and all objects share it. We can access it with a class name and it will be same for every class object.

```python
class Student:
    school = "ABC"
    
    def __init__(self, x, y):
        self.name = x
        self.age = y
        
    def schools(self):
        print(Student.school)  # access it with a class name
        
obj = Student("salman", 19)
obj.schools()
Student.school       # We can also access it outside class
```

```python
# some beginner think that if we write
obj.school="salman"

# Many beginners think this changes the class variable.

# It doesn't.

# Instead, Python creates a new instance variable named school only for s1.
print(Student.school)
print(obj.school)
```

```python
# if we want to change the class attribute we use class name``
Student.school = "xyz"

Student.school
```

# **Static Method**
A method which belongs to the class and can access by class name. It does not need self parameter because it does not belong to object, its belong to class. 

We can also access it by obj but its not recommended. 

In order to make any method static we use `@staticmethod` before the method

```python
class Demo:
    
    @staticmethod
    def show():
        print("salman")
        
Demo.show()
```

### 1 - Write OOP classes to handle the following scenarios:

- A user can create and view 2D coordinates
- A user can find out the distance between 2 coordinates
- A user can find find the distance of a coordinate from origin
- A user can check if a point lies on a given line
- A user can find the distance between a given 2D point and a given line



```python
class Cordinate:
    def __init__(self,x,y):
        self.x_cord=x
        self.y_cord=y
    
    def __str__(self):
        return "<{}, {}>".format(self.x_cord, self.y_cord)
    
    def distance_btw_points(self, other):
        d1 = (other.x_cord-self.x_cord)**2
        d2 = (other.y_cord-self.y_cord)**2
        result = (d1+d2)**0.5
        return result
    
    def distance_from_origin(self):
        d1 = (0-self.x_cord)**2
        d2 = (0-self.y_cord)**2
        result = (d1+d2)**0.5
        return result
        
    def point_lie(self, A, B, C):
        if (A*self.x_cord)+(B*self.y_cord)+C==0:
            print("this point lie on line")
        else:
            print("NOt lie on line")
            
    def distance_from_line(self, A, B, C):
        return abs((A*self.x_cord)+(B*self.y_cord)+C)/(A**2 + B**2)**0.5
        
        
```

```python
cor1 = Cordinate(2, 3)
cor2 = Cordinate(5, 8)

print(cor1)
print(cor2)
```

```python
cor1.distance_btw_points(cor2)
cor1.distance_from_origin()
```

```python
cor1.point_lie(2, -4, 8)
```

```python
cor1.distance_from_line(3, 2, 4)
```
