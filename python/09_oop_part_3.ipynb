# **Super Keyword**
super() means:

Call the parent class.

```python
class Person:
    def __init__(self):
        self.country = "Pakistan"
        
class Student(Person):

    def __init__(self):
        self.roll = 101
        

s = Student()

s.roll    # it work 

s.country   #  but it does not work Because the child's constructor replaces the parent's constructor. This is where super() becomes important.
```

**This is where super() becomes important.**

```python
class Person:
    def __init__(self):
        self.country = "Pakistan"
        
class Student(Person):

    def __init__(self): # first it executed directly while creating object
        super().__init__()  # Then super() also execute the person constructor so in this way the both class constructor initialize
        self.roll = 101
        

s = Student()

s.country
```

```python
class Parent:

    def __init__(self,num):
      self.__num=num

    def get_num(self):
      return self.__num

class Child(Parent):

    def __init__(self,num,val):
      super().__init__(num)
      self.__val=val

    def get_val(self):
      return self.__val

son=Child(100,200)
print(son.get_num())
print(son.get_val())
```

# **Method Overiding**
when the child replaces the parent's implementation.

This is called method overriding.

```python
class Animal:
    def sound(self):
        print("animal sount")
        
class Dog(Animal):
    def sound(self):
        print("Dog bark")
        
        
o = Dog()
o.sound()       # it give me the child sound method what if we need both, the answer is super keyword
```

### **Calling Parent Method After Overriding**


```python
# if we need both method from each class then we will use super keyword

class Animal:
    def sound(self):
        print("animal sount")
        
class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog bark")
        
        
o = Dog()
o.sound() 
```

### **1. Single Inheritance**

Animal

↓

Dog

**One parent.**

**One child.**

```python
class Animal:
    def sound(self):
        print("animal sount")
        
class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog bark")
        
        
o = Dog()
o.sound() 
```

### **2. Multilevel Inheritance**

Animal

↓

Mammal

↓

Dog

**Inheritance continues through multiple levels.**

```python
class Animal:
    def sound(self):
        print("animal sount")

class Mammal(Animal):
    def color(self):
        print("color of mammal")
        
class Dog(Mammal):
    def shape(self):
        print("dog have four legs")
        
o = Dog()
o.sound()    # we can access the all attribute from all classes
o.color()
o.shape()
```

# **Polymorphism**

Polymorphism is the ability of the same interface (method, operator, or function call) to behave differently depending on the object or data it is used with.

# **Method Overriding**
 Method overriding occurs when a child class provides its own implementation of a method that already exists in the parent class.

# **Why Do We Need It?**

Suppose every animal makes a sound.

**Parent class:**



```python
class Animal:

    def sound(self):
        print("Animal makes a sound")
```


Now think about a dog.

Should it print:

Animal makes a sound

No.

A dog should bark.

So the child changes the implementation.

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Bark")


class Cat(Animal):

    def sound(self):
        print("Meow")
        
        
dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

**Same method.**

**Different forms.**

**That's polymorphism.**

Sometimes we want both the parent behavior and the child behavior then we use super().
super() allows the child to call the parent's version before adding its own behavior.


# **Method Overloading**
Method overloading means creating multiple methods with the same name but different parameters.<br>
Python does not allow multiple methods with the same name in a class but it can be implemented in java and other programing languages.

In the below which method survives?

The second one.

The first one is replaced.

So Python doesn't perform traditional method overloading.

```python
class Demo:

    def add(self, a):
        print(a)

    def add(self, a, b):
        print(a + b)
        
ob = Demo()
ob.add(3)
```

```python
ob.add(3, 4)
```

# **Method Overloading vs Method Overriding**

| Method Overloading                      | Method Overriding           |
| --------------------------------------- | --------------------------- |
| Same class                              | Parent and child classes    |
| Same method name                        | Same method name            |
| Different parameters                    | Same parameters (typically) |
| Python doesn't support it traditionally | Fully supported in Python   |

# **Abstraction**
Abstraction is the process of hiding the implementation details and showing only the essential features to the user.

**In simple words:**

The user should know what an object can do, but not necessarily how it does it.

### **Abstract class**
An abstract class is a class that cannot be instantiated directly.

shape = Shape()

If Shape is abstract, this is not allowed.

Why?

Because the abstract class is only a blueprint.

### **Abstract Method**

An abstract method is a method that has:

A name
Parameters

But no implementation

**It says:**

Every child class must implement this method.

# **Creating an Abstract Class**
For creating an Abstract Class we should import the ABC and abstractmethod from abc

```python
from abc import ABC, abstractmethod

class Animal(ABC):  # this tell python that Animal is an abstract class

    @abstractmethod         # this decorator tell python that every child class must have implement this method.
    def sound(self):
        pass
```

```python
class Dog(Animal):
    def sound(self):
        print("hi")
        
dog = Dog()
dog.sound()     # it does not give en error because we have implement the sound method in the child class
```

```python
class Cat(Animal):
    def name(self):
        print("hi")
        
cat = Cat()     # it give an error because we have not implement the sound method 
```

# **Another Example**

```python
class CoffeeMachine:

    def make_coffee(self):
        self.__boil_water()
        self.__grind_beans()
        self.__mix()
        print("Coffee is Ready!")

    def __boil_water(self):
        print("Boiling Water")

    def __grind_beans(self):
        print("Grinding Beans")

    def __mix(self):
        print("Mixing Coffee")
```

```python
coffee = CoffeeMachine()

coffee.make_coffee()
```

This is Abstraction Because the user only knows:

coffee.make_coffee()

**The user doesn't need to know:**

- How water is boiled
- How beans are ground
- How coffee is mixed

The implementation is hidden.
