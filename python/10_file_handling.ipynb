### **File Handling**
File Handling is the process of creating, opening, reading, writing, updating, and deleting files stored on a computer using a programming language.

**In simple words:**

File handling allows a program to permanently store and retrieve data from files

### **I/O**

I/O stands for Input and Output.

I = Input

O = Output

# **File I/O**

When working with files:

### **File Input**

Reading data from a file into your program.

### **File Output**

Writing data from your program into a file.


# **Types of Files**

There are mainly two types of files:

- Text Files
- Binary Files

### **1. Text File**

A text file stores data as human-readable characters. examples .txt, .csv, .json

### **1. Binary File**

A binary file stores data in binary format (bytes) instead of readable text.

Humans generally cannot understand the contents directly. examples .jpg, .mp3, .pdf, .db


# **Opening Files (open())**
open() is a built-in Python function used to open a file so that your program can read from it or write to it.

**Syntax :**
open(file, mode)

**1. file**

The path to the file.

**2. mode**

Tells Python what you want to do

```python
f = open("sample.txt", "w")
f.write("This file is just for file handling practice\n Hi how are you\nThis is new lines")
f.close()
```

```python
f = open("sample.txt", "r")
print(f.read())
f.close()
```

# **File Modes**

The mode tells Python how you want to interact with the file.

### **Read Mode (r)**
Read an existing file.

```python
f = open("sample.txt", "r") # the r is for read mode as we open the file in a read mode
print(f.read())
f.close()
```

### **Write Mode (w)**

Write data to a file.If the file already contains:
and you open it in "w" mode, the existing content is erased immediately.

```python
f = open("sample.txt", "w")
f.write("This file is over write")
f.close()
```

### **Append Mode (a)**

Add new data to the end of the file.

```python
f = open("sample.txt", "a")
f.write(" The text will be append at the end")
f.close()
```

```python
f = open("sample.txt", "a")
f.write("\nThis is for another line ")
f.close()
```

### **Exclusive Creation Mode (x)**


Create a new file. If file exsist it will raise error.

```python
f = open("sample1.txt", "x")
f.write("ho")
f.close()
```

### **Binary Read Mode (rb)**

Read binary files. 

```python
f = open("img.jpg", "r")
f.read()
f.close()       # it raise an error which mean we can read or write any binary file with simple read we should use read binary
```

```python
f = open("img.jpg", "rb")
f.read()                # we can read the binary file with read binary
f.close()
```

### **Binary Write Mode (wb)**

Write binary data.

```python
# copying the Image with Method - 1
f = open("img.jpg", "rb")
im = f.read()
f.close()
f = open("copy.jpg", "wb")
f.write(im)
f.close()


# copying the Image with Method - 2
f = open("img.jpg", "rb")
p=open("copy.jpg", "wb")
p.write(f.read())
p.close()
f.close()
```

| `a`  | ❌ No             | ✅ Yes         | ❌ No                   | Append          |
| `x`  | ❌ Must not exist | ✅ Yes         | ❌ No                   | Create new file |

# **Reading Files**

Once a file is opened in read mode, you can retrieve its contents using different methods.

### **read()**
Reads data from the file.

**Syntax :**
file.read(size=-1)


size (optional): Number of characters (or bytes in binary mode) to read.<br>
Default -1 means read everything.

```python
f = open("sample.txt", "r")
print(f.read())
f.close()
```

```python
f = open("sample.txt", "r")
print(f.read(5), end="")
print(f.read(5), end="")
f.close()
```

### **readline()**
Reads one line at a time.

```python
f = open("sample.txt", "r")
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())
f.close()
```

### **readlines()**
Reads all lines and returns them as a list of strings.

```python
f = open("sample.txt", "r")
print(f.readlines())
f.close()
```

### **Summary**

| Method        | Returns                   |
| ------------- | ------------------------- |
| `read()`      | Entire file as one string |
| `readline()`  | One line as a string      |
| `readlines()` | List containing all lines |

### **tell()**
tell() returns the current position of the file pointer.


```python
with open("sample.txt", "w") as file:
    file.writelines(l)
```

```python
with open("sample.txt", "r") as file:
    print(file.read(5))
    print(file.tell())
```

```python
with open("sample.txt", "r") as file:
    print(file.read(10))
    print(file.tell())
```

### **seek()**

seek() moves the file pointer to a specified position.

```python
with open("sample.txt", "r") as file:
    print(file.read(10))
    print(file.seek(0))
    print(file.read(10))
```

```python
with open("sample.txt", "w") as file:
    file.write("Hi how are you.")
    file.seek(0)
    file.write("okay")
```

```python
with open("sample.txt", "r") as file:
    print(file.read(10))
    file.seek(0)
    print(file.read())
```

# **tell() Vs seek()**

| tell()                  | seek()                                                              |
| ----------------------- | ------------------------------------------------------------------- |
| Shows pointer position  | Moves pointer                                                       |
| Doesn't change location | Changes location                                                    |
| Returns an integer      | Returns the new position (or `None` in older versions) after moving |

# **JSON**
JSON (JavaScript Object Notation) is a lightweight text-based format used to store and exchange data between programs.

**In simple words:**

JSON is a universal language for sharing structured data between different applications and programming languages.

**How can another program understand a Python dictionary?**

A Java program cannot directly read a Python dictionary.

A JavaScript program cannot directly understand Python objects.

So we need a common format.

That common format is JSON.

### **JSON Module**
The json module is a built-in Python module used to convert Python objects to JSON and JSON back to Python objects.

It performs two main operations:

Serialization
Deserialization

These are the most important concepts.

Note : With the help of json module we can only serlialize and deserialize the built-in data types not the user define data types(class).For storing the object we need to use pickling.

```python
import json
```

```python
with open("sample.json", "w") as file:
    json.dump(dic, file)
    
```

```python
with open("sample.json", "r") as file:
    data = json.load(file)
    
print(data)
print(type(data))
```

```python
with open("sample.json", "w") as file:
    json.dump(5, file)

with open("sample.json", "r") as file:
    data = json.load(file)
    
print(data+2)
```

### **Serialization**
Serialization is the process of converting a Python object into JSON format so it can be stored or transmitted.

```python
with open("sample.json", "w") as file:
    json.dump(dic, file, indent=4)    
```

### **Deserialization**

Deserialization is the process of converting JSON back into Python objects.

```python
with open("sample.json", "r") as file:
    data = json.load(file)
    
print(data)
print(type(data))
```

# **Serialization vs Deserialization**

| Serialization     | Deserialization      |
| ----------------- | -------------------- |
| Python → JSON     | JSON → Python        |
| Save or send data | Read or receive data |
| Export            | Import               |


# **The Four Main Functions of the json Module**

### **dump()**
Writes a Python object directly into a JSON file.

```python
l = [1, 2, 3, 4, 5, 6]

with open("sample.json", "w") as file:
    json.dump(l , file)
```

### **load()**

Reads JSON data from a file and converts it into a Python object.

```python
with open("sample.json", "r") as file:
    d = json.load(file)
    
print(d)
print(type(d))
```

**JSON does not know how to convert a custom Python object into JSON.**

**That's why the default parameter exists.**

```python

class Person:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks


person = Person("salman", 34, 344)

def show(obj):
    if isinstance(obj, Person):
        return {"name":obj.name, "age":obj.age, "marks":obj.marks}



with open("sample.json", "w") as file:
    json.dump(person, file, default=show)
    
    
with open("sample.json", "r") as file:
    data = json.load(file)
    

print(data)
print(type(data))    
    
```

### **dumps()**

Notice the extra s.

dumps() does not create a file.

It converts a Python object into a JSON string.

### **loads()**

Again, notice the extra s.

It converts a JSON string into a Python object.

don't memorize these functions by their names. Instead, ask yourself one question:

**"Is my JSON inside a file, or is it inside a string?"**

If it's inside a file → use dump() or load().

If it's inside a string → use dumps() or loads().

# **Pickling**
Pickling is the process of converting a Python object into a byte stream so that it can be stored in a file or transmitted over a network.

**Simple definition:**

Python Object → Byte Stream

# **Unpickling**
Unpickling is the process of converting the stored byte stream back into the original Python object.

Simple definition:

Byte Stream → Python Object

# **Pickle**
Pickle is a built-in Python module that is used to serialize and deserialize Python objects.

In simple words:

Pickle converts Python objects into bytes so they can be stored or transmitted, and later converts those bytes back into the original Python objects.

```python
import pickle 
```

### **Pickle.dump()**

### **Pickle.load()**

```python
import pickle as pkl

class Demo:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks
        
    def display(self):
        return f"My name is {self.name} and age is {self.age} and got {self.marks} marks in matriculation"

obj = Demo("salman", 19, "117 out of 1200")

    
with open("sample.pkl", "wb") as file:
    pkl.dump(obj, file)
    
with open("sample.pkl", "rb") as file:
    objs = pkl.load(file)


objs.name
```

```python
print(objs.display())
```

```python
objs.name = "muhammad"
print(objs.display())
```

### **1. Write a program which read the file chunk by chunk not whole at a time.**

```python
l = [str(f"{i}\n") for i in range(1, 1001)]

with open("sample3.txt", "w") as file:
    file.writelines(l)
```

```python
# the problem is that if we read the file at once then it will use our memory and then load to the python instead we will read some part of file then clean the memory and then some part then clean the memory.

with open("sample3.txt", "r") as file:
    chunk = 20
    data = file.read(chunk)
    while len(data)>0:
        print(data, end='')
        data=file.read(chunk)
```

### **2. Use for loop to load the txt file line by line untill it reach to end**

```python
with open("sample3.txt", "r") as file:
    data = file.readline()
    while True:
        if len(data)>0:
            print(data, end="")
            data = file.readline()
        else:
            break
```
