Python errors are mainly divided into two categories.
1. Syntax Errors
2. Exceptions (Runtime Errors)

### **2. Exceptions (Runtime Errors)**

An Exception is an error that occurs while the program is running.

Notice the difference.

**Syntax Error:**

Program cannot start.

Exception:

Program starts, but something goes wrong later.

````python
print("start")
print(10/0)
print("end")
````

### **Differnce between Syntax and Exception Error**

| Syntax Error                         | Exception                         |
| ------------------------------------ | --------------------------------- |
| Happens before execution             | Happens during execution          |
| Program doesn't start                | Program starts, then crashes      |
| Caused by invalid Python syntax      | Caused by unexpected situations   |
| Cannot be handled using `try-except` | Can be handled using `try-except` |

# **Exception Handling**
Exception Handling is a mechanism that allows a program to handle runtime errors gracefully instead of crashing.

**Simple definition:**

Exception Handling lets your program continue running even when an error occurs.

# **Four Keywords of Exception Handling**

**1. try**<br>
**2. except**<br>
**3. else**<br>
**4. finaly**<br>


### **try**
The try block contains code that might produce an exception.

**Think of it as saying:**

"Python, this code may cause an error. Please watch it carefully."


<br>


### **except**

The except block runs only if an exception occurs inside the try block.

**Think of it as:**

"If something goes wrong, execute this code instead of crashing."

```
try:
    risky code

except:
    handle error
```

````python
try:
    print(10/0)
except:
    print("error")
````

````python
# if python got error inside try block and after that Python stops executing the remaining code inside the try block.
try:
    print("a")
    print(10/0)
    print("b")
except:
    print("error")
print("end")
````

### **Catch Specific Exceptions**

````python

try:
    a = int(input("enter no."))
    b = int(input("enter no."))
    print(a/b)
except ValueError:
    print("Please input valid number")
except ZeroDivisionError:
    print("The number cannot divided by Zero")
````

### **One except Handle Multiple Errors**

````python

try:
    a = int(input("enter no."))
    b = int(input("enter no."))
    print(a/b)
except (ValueError, ZeroDivisionError):
    print("Please input valid number")
    

````

### **else**
Many beginners misunderstand else.

It is not related to if.

In exception handling,

**else means:**

Run this block only if NO exception occurs.

````python
try:
    a = int(input("enter no."))
    b = int(input("enter no."))
    result = a/b
except ValueError:
    print("Please input valid number")
except ZeroDivisionError:
    print("The number cannot divided by Zero")
else:
    print(result)
````

### **finally**

The finally block always executes, whether an exception occurs or not.

**Think of it as:**

"No matter what happens, execute this code."

````python
try:
    a = int(input("enter no."))
    b = int(input("enter no."))
    result = a/b
except ValueError:
    print("Please input valid number")
except ZeroDivisionError:
    print("The number cannot divided by Zero")
else:
    print(result)
finally:
    print("Thank")
````

# **Catching Exception**


````python
def add(a,b):
    if type(a)==int and type(b)==int:
        return a+b
    else:
        raise Exception("The arguments should be integers")
    
try:
    result = add(4, 3)
except Exception as e:
    print(e)
else:
    print(result)
finally:
    print("Calculation Don")
    
````

# **Custom Exception**
A custom exception is an exception class created by the programmer.

Since exceptions are classes,

you can create your own class.

A custom exception should inherit from the Exception class.

````python
class OperationCantPerforom(Exception):
        pass

def add(a,b):
    if type(a)==int and type(b)==int:
        return a+b
    else:
        raise OperationCantPerforom("data type should be int")

try:
        add(3, "salman")
except OperationCantPerforom as e:
        print(e)
````

````python
class IncorrectPassword(Exception):
    def __init__(self):
        print("Incorrect password")
    
    def later(self):
        print("Please try again later after 1 hour")


def login(email, password):
    if email=="salman@gmail.com" and password==12345:
        print("successfully login")
    else:
        raise IncorrectPassword
    
try:
    result = login("salman@gmail.com", 212345)
except IncorrectPassword as e:
    e.later()
except Exception as e:
    print(e)
````
