# **Namesspaces**

A namespace is a container that stores the relationship between names (identifiers) and objects.

**In simple words:**

A namespace is like a dictionary where Python stores variable names and the objects they refer to.

### **Why Do We Need Namespaces?**

Imagine Python had no namespaces.
```
x = 10

def show():
    x = 20
```
Now there are two variables named x.

Which one should Python use?

It wouldn't know.

Namespaces solve this problem by keeping them in different containers.

### **Types of Namespaces**

Python mainly has four namespaces.

Namespaces

│

├── Built-in Namespace

├── Global Namespace

├── Enclosing Namespace

└── Local Namespace

These four correspond to the LEGB Rule

### **1. Built-in Namespace (B)**

This namespace is created automatically when Python starts.

It contains everything Python already provides.

The Global Namespace belongs to the entire Python file.

Everything created outside functions belongs here.

````python
a = "salman"  # global
def show():
    a = "muhammad"   # local
    
print(a)
````

### **3. Local Namespace (L)**

Every time a function is called,

Python creates a brand new namespace for that function.

````python
def show():
    a = "salman" # local
    print(a)
    
show()
````

### **4. Enclosing Namespace (E)**

This one confuses many beginners.

It appears when one function is inside another function.

````python
def show():
    
    a = "salman"   # enclosing namesspace
    def name():
        print("hi this is name")         # local namesspace
    print("this is show")
    return name()
````

# **Decorator**
A decorator is a function that takes another function as input, adds some extra functionality to it, and returns a new function without changing the original function's code.

In simple words:

A decorator wraps a function to add new behavior before or after it runs.

Think of it as adding a protective or decorative layer around an existing function.


````python
def Hello():
    print("slaman")
    return "this si s"
    
def decorator(fun):
    def wrapper():
        print("****")
        fun()
        print("****")
    return wrapper


s = decorator(Hello)
s()

````

````python

def add(a, b):
    return a+b

def decorator(fun):
    def wrapper(x, y):
        print("started")
        result = fun(x, y)
        print("hi how are you")
        return result
    return wrapper

s = decorator(add)
s(3, 4)
````

# **Symbol @**
This Symbol(@)  is just a shortcut to use the decorator:

````python
def decorator(fun):
    def wrapper():
        print("****")
        fun()
        print("****")
    return wrapper


@decorator             # this is the shortcut to use decorator with function
def hello():
    print("salman")
    
hello()
````

### **Timer Decorator**

````python
import time

def timer(func):
    def wrapper(*args):
        start = time.time()
        func(*args)
        print(func.__name__, "take", start-time.time(), "secs")
    return wrapper
        
        
@timer
def add(a, b):
    print(a+b)
    
@timer  
def sqr(num):
    print(num**2)

@timer
def cube(num):
    print(num**3)

@timer
def check():
    time.sleep(4)
    print("how are you")
    
     
try:
    check()
except Exception as e:
    print(e)
````

### **Datatype Check**

````python
def typechecker(datatype):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            for i in args:
                if type(i)!=datatype:
                    raise ValueError("Please input valid datatype")
            func(*args)
        return inner_wrapper
    return outer_wrapper
            
@typechecker(int)
def add(a, b):
    print(a+b)
    
    
@typechecker(str)
def detail(name):
    print("your name is",name)
    
@typechecker(str)
def introduc(name, age, marks, classs):
    print(f"My name is {name} {age} year with {marks} marks in class {classs}")
    
try:
    detail("salman")
except Exception as e:
    print(e)
    
````

<b><h1><center>Recursion</center></h1></b>

# **Recursion**

Recursion is a programming technique in which a function calls itself to solve a problem.

**Simple definition:**

A recursive function is a function that solves a problem by calling itself on a smaller version of the same problem until it reaches a stopping condition.


### **Two Essential Parts of Every Recursive Function**

Every recursive function must have:

- Base Case
- Recursive Case

Without these, recursion will never work correctly.

### **1. Base Case**
The base case is the condition where recursion stops.

Think of it as:

"I have reached the smallest possible problem. I don't need to call myself anymore."

### **2. Recursive Case**

The recursive case is where the function calls itself.

````python
def count(n):
    if n==0:
        print("finished")
        return

    print(n)
    count(n-1)
    
a = count(3)
````

````python
def countup(n):
    if n==0:
        return  
    countup(n-1)      
    print(n)
    

countup(4)
````

````python
def sumofno(num):
    if num==0:
        return 0
    
    return num + sumofno(num-1)

print(sumofno(6))
````

````python
def sumofdigit(num):
    if num==0:
        return 0
    
    return (num%10)+sumofdigit(num//10)


sumofdigit(1234567)
````

# **Iteration**
Iteration is the process of accessing one item at a time from a collection until all items have been processed.

**Simple definition:**

Iteration means visiting every element one by one.


# **Iterable**

An iterable is an object that can be iterated over.

# **Iterator**

An iterator is an object that remembers its current position and returns the next value whenever asked.

Simple definition:

An iterator is an object that gives one item at a time and remembers where it stopped.

when we apply dir on iterator then it will contian iter and also next.

````python
# to check whether an object contain iter or not then we use the dir magic method
a = [3, 4, 5, 7]
dir(a)
````

````python
# function to check iterator and iterable

def check(obj):
    if "__iter__" in dir(obj) and "__next__" in dir(obj):
        print(type(obj), "is iterator")
    elif "__iter__" in dir(obj):
        print(type(obj), "is iterable")
    else:
        print(type(obj), "Not iterable nor iterator")
        
        
a = 8
check(a)

b = [2, 3, 4, 5]
check(b)

c = iter(b) # iter function will convert iterable to iter and it will contain the next magic method as we can apply next function

check(c)
````

### **Own For Loop**

````python
def ownforloop(iterable):
    iterator=iter(iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
        
        
        
a = 'salman khan'
ownforloop(a)
````

````python
l = [3, 4, 5, 6, 7, 8, 9]
ownforloop(l)
````

### **Generator**

A generator is a special type of iterator created using the yield keyword.

**Simple definition:**

A generator produces values one at a time, only when needed.


**A generator saves memory**

A generator is memory-efficient because it doesn't create all results at once. If the values it produces are themselves very large objects, those objects still occupy memory while they exist. The key advantage is that the generator creates values on demand instead of storing the entire sequence in memory at the same time.

````python
def generator():
    yield "salman"
    yield "naveed"
    yield "ubaid"
    yield "nadeem"
    

gen = generator()
print(next(gen))

for i in gen:
    print(i)
````

````python
def gene(start, end):
    for i in range(start, end):
        yield i
        

   
for i in gene(1, 5):
    print(i)
    
````

````python
gen = gene(1, 4)
next(gen)
next(gen)
next(gen)
````

````python
genes = gene(1, 10)

for i in genes:
    print(i)
````
