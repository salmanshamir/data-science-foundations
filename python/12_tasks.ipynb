### `Q3:` Write a recursive program to to calculate `gcd` and print no. of function calls taken to find the solution.
```
gcd(5,10) -> result in 5 as gcd and function call 4
```

````python
#Write your code here
count = 0
def gcd(a, b):
    global count
    count +=1
    if b==0:
        return a
    return gcd(b, a%b)


result = gcd(16, 24)

print(f"gcd(16, 24) -> result {result} and function call {count}")
````

Create your own `MyEnumerate` function such that someone can use it instead of enumerate. It will need to return a `tuple` with each iteration, with the first element in the tuple being the `index` (starting with 0) and the second element being the `current element` from the underlying data structure. Trying to use `MyEnumerate` with a noniterable argument will result in an error.

```
for index, letter in MyEnumerate('abc'):
    print(f'{index} : {letter}')
```

Output:
```
0 : a

```

````python
#Write your code here

def MyEnumerate(iterable):
    if '__iter__' in dir(iterable):
        iterator = iter(iterable)
        index = 0
        while True:
            try:
                element = next(iterator)
                yield index, element
                index+=1
            except StopIteration:
                break
    else:
        raise TypeError(f"{iterable} object is not iterable")
        
            
        
a = ["salman", "khan"]

b = "this is just for texting purpose"    
    
for i,j in MyEnumerate(a):
    print(i, j)


````

### `Q6:` Generator time elapsed
Write a generator function whose argument must be iterable. With each iteration, the generator will return a two-element tuple. The first element in the tuple will be an integer indicating how many seconds have passed since the previous iteration. The tuple’s second element will be the next item from the passed argument.

Note that the timing should be relative to the previous iteration, not when the
generator was first created or invoked. Thus the timing number in the first iteration
will be 0

```
for t in elapsed_since('abcd'):
    print(t)
    time.sleep(2)
```

Output:
```
(0.0, 'a')
(2.005651817999933, 'b')
(2.0023095009998997, 'c')
(2.001949742000079, 'd')
```
Note: Your output may differ because of diffrent system has different processing configuration.

````python
#Write yor code

import time
def times(iterable):
    if "__iter__" in dir(iterable):
        iterator = iter(iterable)
        previous = None
        while True:
            try:
                element = next(iterator)
                if previous==None:
                    elapsed = 0.0
                else:
                    elapsed = time.time()-previous
                previous = time.time()
                yield (round(elapsed, 2), element)
            except StopIteration:
                break
    else:
        raise TypeError("Please input iterable")
    
    
    
a = [1, 2, 3, 4, 5]

for i in times(a):
    print(i)
    time.sleep(2)
       
       
        
````

## Decorators

### `Q7:` Write a Python program to make a chain of function decorators (bold, italic, underline etc.) on a given function which prints "hello world"

```
def hello():
    return "hello world"
```

```
bold - wrap string with <b> tag. <b>Str</b>
italic - wrap string with <i> tag. <i>Str</i>
underline- wrap string with <u> tag. <u>Str</u>
```

````python
#Write your code here


def decorator(fun):
    funct = fun
    def bold():
        bolded = f"<b>{funct()}</b>"
        def italic():
            italiced = f"<i>{bolded}</i>"
            def underline():
                under = f"<u>{italiced}</u>"
                return under            
            return underline()
        return italic()
    return bold

@decorator
def hello():
    return "Hello World"




def bold(fun):
    def wrapper():
        bolded = f"<b>{fun()}</b>"
        return bolded
    return wrapper

def italic(fun):
    def wrapper():
        italiced = f"<i>{fun()}</i>"
        return italiced
    return wrapper

def underline(fun):
    def wrapper():
        under = f"<u>{fun()}</u>"
        return under 
    return wrapper

@underline
@italic
@bold
def nam():
    return "salman"

nam()
hello()
````

### `Q8:` Write a decorator called `printer` which causes any decorated function to print their return values. If the return value of a given function is `None`, printer should do nothing.



````python
def printer(fun):
    def wrapper():
        result = fun()          
        if result is not None:  
            print(result)

    return wrapper


@printer
def hello():
    return "Hello World"


@printer
def add():
    return 100


@printer
def greet():
    print("Hi Salman")


hello()
add()
greet()
````

### `Q9:` Make a decorator which calls a given function twice. You can assume the functions don't return anything important, but they may take arguments.
```
#Lets say given function
def hello(string):
    print(string)

#on calling after specified decorator is inplaced
hello('hello')
```

Output:
```
hello
hello
```

````python
#Write your cod here
def decorator(fun):
    def wrapper(*args):
        fun(*args)
        fun(*args)
    return wrapper

@decorator
def hello(string):
    print(string)
    
hello("salman")
````

### `Q10:` Write a decorator which doubles the return value of any function. And test that decoratos is working correctly or not using `asert`.

```
add(2,3) -> result in 10. Without decorator it should be 5.
```

````python
# Write your code here

def decorator(fun):
    def wrapper(a, b):
        re = fun(a, b)
        return re*2
    return wrapper
    
            
@decorator
def add(a, b):
    return a+b



assert add(4, 5) == (4+5)*2

add(2, 4)
````
