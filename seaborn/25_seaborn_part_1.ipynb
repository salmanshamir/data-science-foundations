# **Seaborn**

Seaborn is a Python library for statistical data visualization. It is built on top of Matplotlib and is designed to work especially well with Pandas DataFrames.



| Matplotlib                      | Seaborn                           |
| ------------------------------- | --------------------------------- |
| General-purpose visualization   | Statistical visualization         |
| Lower-level                     | Higher-level                      |
| More control                    | Easier syntax                     |
| Works with many data sources    | Works especially well with Pandas |
| More code for statistical plots | Less code                         |
| Basic/default appearance        | Better default appearance         |


Seaborn does not replace Matplotlib.

Internally, Seaborn uses Matplotlib for rendering plots.


# **Types of seaborn plots**
Seaborn plotting functions are broadly divided into:

- Axes-level functions
- Figure-level functions

### **Axis level Plots**
An axes-level function creates a plot on one particular Axes.

### **Figure level Plots**
Figure-level functions work at the Figure level and can manage the overall figure, including creating multiple Axes when necessary.

                 Figure
                   │
          ┌────────┴────────┐
          │                 │
        Axes 1            Axes 2
          │                 │
       Plot 1             Plot 2



````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns

df = sns.load_dataset('tips')
````

<br>

# **Relationship Plots**
Relationship plots are used to understand the relationship between two or more variables.

**There are two main relationship plots**
- scatterplot
- lineplot

### **Scatter Plot**
A scatter plot displays individual observations as points.

### **Line Plot**
A line plot connects observations with lines.


```
Relationship
│
├── Axes-level
│   ├── scatterplot()
│   └── lineplot()
│
└── Figure-level
    └── relplot()
        ├── scatter
        └── line
```


<br>

### **Scatter Plot - Axes Level**

````python
df.columns
````

````python
sns.scatterplot(data=df, x="tip", y="total_bill", hue='sex', style='day', palette=['red', 'green'], size='size')
plt.show()
````

<br>

### **Scatter Plot - Figure Level**

````python
sns.relplot(data=df, x='tip', y='total_bill', kind='scatter', hue='sex', style='smoker')
plt.show()
````

<br>

### **Line Plot - Axes Level**

````python
gap = px.data.gapminder()
temp_df = gap[gap['country'].isin(["India", "Pakistan"])]
sns.lineplot(data=temp_df, x="year", y='lifeExp', hue='country')
plt.show()
````

````python
sns.relplot(data=gap[gap["country"].isin(["India", 'Pakistan', "Brazil", "Germany"])], x="year", y="lifeExp", hue='country', style='continent', kind='line')
plt.show()
````

<br>

### **Line Plot - Figure Level**

````python
sns.relplot(data=temp_df, x="year", y='lifeExp', hue='country', kind='line')
plt.show()
````

# **Distribution Plots**
A distribution plot is used to understand how the values of a variable are distributed.

We may want to know:

Where are most values concentrated?
What is the spread of the data?
Is the distribution symmetric?
Is it skewed?
Are there unusual/extreme values?
Is there one peak or multiple peaks?

These are questions about the distribution of a variable.

**There are main three distribution plots**
- histogram
- kdeplot
- rugplot


### **Histogram**
A histogram divides the numerical range into intervals called bins and counts how many observations fall into each bin.

### **KDE Plot**
KDE = Kernel Density Estimation

KDE estimates the probability density of the data and represents it as a smooth curve.

### **Rug Plot**
A rug plot represents each individual observation as a small line along an axis.

<br>

### **Histogram - Axis Level**

````python
sns.histplot(data=df, x='total_bill')
plt.show()
````

````python
sns.histplot(data=df, x='total_bill', hue='sex', kde=True)
plt.show()
````

````python
sns.histplot(data=df, x='total_bill', hue='sex', element='step')
plt.show()
````

````python

sns.histplot(data=df, x='total_bill', y='tip')
plt.show()
````

<br>

### **Histogram - Figure Level**

````python
sns.displot(data=df, x='total_bill', kind='hist', height=5, aspect=1.5)
plt.show()
````

````python
sns.displot(data=df, x='total_bill', kind='hist', height=5, aspect=1.5, hue='sex', element='step')
plt.show()
````

````python
sns.displot(data=df, x='total_bill', kind='hist', height=5, aspect=1.5, kde=True)
plt.show()
````

<br>

### **KDE Plot - Axis Level**

````python
sns.kdeplot(data=df, x='total_bill')
plt.show()
````

````python
sns.kdeplot(data=df, x='total_bill', hue='sex', palette={'Male':'red', 'Female':'blue'})
plt.show()
````

````python
sns.kdeplot(data=df, x='total_bill', hue='sex', fill=True, palette=["green", 'red'])
plt.show()
````

````python
sns.kdeplot(data=df, x='total_bill', y='tip')
plt.show()
````

<br>


### **KDE Plot - Figure Level**

````python
sns.displot(data=df, x='total_bill', kind='kde', fill=True)
plt.show()
````

<br>

### **Rug Plot - Axis Level**

````python
sns.rugplot(data=df, x='total_bill')
plt.show()
````

````python
# generally we plot rug with ked

sns.rugplot(data=df, x='total_bill')
sns.kdeplot(data=df, x='total_bill', color='skyblue')
plt.show()
````

# **Matrix Plot**

Matrix plot is used to visually disply a matrix

**The main type of matrix plot is :**
- Heatmap

### **Heatmap**
Heatmap is an axis level plot as it does not have figure level and is used for matrix.

````python
tamp = gap[gap["continent"]=='Asia']
p = tamp.pivot(index='country', columns='year', values='lifeExp')

sns.heatmap(p)
````

````python
tamp2 = gap[gap["continent"]=='Europe']
plt.figure(figsize=(10, 10))
sns.heatmap(tamp2.pivot(index='country', columns='year', values='lifeExp'), annot=True, cmap='viridis')
plt.show()
````
