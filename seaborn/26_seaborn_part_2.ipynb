### **Categorical Plot**
A categorical variable contains categories rather than continuous numerical values.

- Stripplot
- Swarmplot
- Boxplot
- Violinplot
- Barplot
- Countplot

### **Strip Plot**
A strip plot displays individual observations for each category.

### **Swarm Plot**
A swarm plot is similar to a strip plot, but it intelligently positions points to avoid overlapping.

### **Box Plot**
A box plot summarizes a numerical distribution using important statistics.

### **Violin Plot**
 A violin plot combines ideas from : box plot + KDE/distribution plot

 ### **Bar Plot**
 A bar plot compares a numerical statistic across categories.

 ### **Count**
 A count plot is used to show how many observations belong to each category.

# **Facet Grid**
FacetGrid is a Seaborn class used to create a grid of multiple subplots based on categorical variables.
It work like facet plot of figure level plots but it is more customizable than facet plot.

```python
g = sns.FacetGrid(data=df, col='sex', row='day', hue='smoker')
g.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```

```python
g = sns.FacetGrid(data=df, col='sex', row='day')
g.map(sns.countplot, 'smoker')
plt.show()
```

Pair plot = visualization of pairwise relationships among multiple variables.

```python
sns.pairplot(data=df)
plt.show()
```

```python
sns.pairplot(data=df, hue='sex')
plt.show()
```

```python
sns.pairplot(data=df, vars=["total_bill", "tip"])
plt.show()
```

```python
sns.pairplot(data=df, vars=["total_bill", "tip"], hue='sex')
plt.show()
```

```python
sns.pairplot(data=df, vars=["total_bill", "tip"], hue='sex', kind='reg')
plt.show()
```

```python
sns.pairplot(data=df, vars=["total_bill", "tip"], hue='sex', kind='hist', diag_kind='kde')
plt.show()
```

```python
sns.pairplot(data=df, vars=["total_bill", "tip"], hue='sex', kind='kde', diag_kind='hist')
plt.show()
```

<br>

### **Pair Grid**

pairplot() is a convenient high-level function, while PairGrid gives you control over each part of the pairwise plot matrix.

It creates a grid of Axes for pairwise relationships between variables, but unlike pairplot(), it does not automatically decide exactly what plot should go where.

You create the grid first, and then tell Seaborn:

- what to put on the diagonal
- what to put on the upper triangle
- what to put on the lower triangle

```python
g = sns.PairGrid(data=df)
g.map(sns.kdeplot)
plt.show()
```

```python
g = sns.PairGrid(data=df, hue='sex')
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.scatterplot)
plt.show()
```

```python
g = sns.PairGrid(data=df, hue='sex')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
plt.show()
```

```python
g = sns.PairGrid(data=df, hue='sex')
g.map_diag(sns.histplot)
g.map_lower(sns.scatterplot)
g.map_upper(sns.histplot)
plt.show()
```

### **Joint Plot**
A joint plot studies the relationship between exactly TWO variables, while also showing the individual distribution of each variable.

```python
sns.jointplot(data=df, x='total_bill', y='tip')
plt.show()
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', hue='sex')
plt.show()
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='kde')
plt.show()
```

```python
sns.jointplot(data=df, x='total_bill', y='tip', kind='reg')
plt.show()
```

### **Joint Grid**

```python
g = sns.JointGrid(data=df, x='total_bill', y='tip', hue='sex')
g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.kdeplot)
```

```python
g = sns.JointGrid(data=df, x='total_bill', y='tip', hue='sex')
g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.histplot)
```

```python
g = sns.JointGrid(data=df, x='total_bill', y='tip', hue='sex')
g.plot_joint(sns.lineplot)
g.plot_marginals(sns.histplot)
```
