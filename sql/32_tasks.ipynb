### Perform the following queries

1. Show records of 'male' patient from 'southwest' region.



```python
pd.read_sql_query("SELECT * FROM insurance where gender='male' AND region='southeast'", con=con)
```

2. Show all records having bmi in range 30 to 45 both inclusive.


```python
pd.read_sql_query("SELECT * FROM insurance where bmi BETWEEN 30 AND 45", con=con)
```



3. Show minimum and maximum bloodpressure of diabetic patient who smokes. Make column names as MinBP and MaxBP respectively.


```python
pd.read_sql_query("SELECT MAX(bloodpressure) AS 'MaxBP', MIN(bloodpressure) AS 'MinBP' FROM insurance WHERE diabetic='Yes' AND smoker='Yes'", con=con)
```

4. Find no of unique patients who are not from southwest region.


```python
pd.read_sql_query("SELECT COUNT(DISTINCT(PatientID)) AS 'Patient' FROM insurance where region!='southwest'", con=con)
```

5. Total claim amount from male smoker.


```python
pd.read_sql_query("SELECT SUM(claim) AS 'Total_claim' FROM insurance WHERE smoker='Yes'", con=con)
```


6. Select all records of south region.


```python
pd.read_sql_query('SELECT * FROM insurance WHERE region LIKE "south%"', con=con)
```

7. No of patient having normal blood pressure. Normal range[90-120]


```python
pd.read_sql_query("SELECT COUNT(*) AS 'patient_no' FROM insurance WHERE bloodpressure BETWEEN 90 AND 120", con=con)
```

8. No of pateint belo 17 years of age having normal blood pressure as per below formula -
    - BP normal range = 80+(age in years × 2) to 100 + (age in years × 2)

    - Note: Formula taken just for practice, don't take in real sense.


```python
pd.read_sql_query("SELECT COUNT(*) FROM insurance WHERE age < 17 AND bloodpressure >= 80 + (age * 2) AND bloodpressure <= 100 + (age * 2)", con=con)
```

9. What is the average claim amount for non-smoking female patients who are diabetic?


```python
pd.read_sql_query("SELECT AVG(claim) AS 'avg_claim' FROM insurance WHERE smoker='No' AND gender='female' AND diabetic='Yes'", con=con)
```
