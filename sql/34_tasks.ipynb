### `Q-1` Find top-5 most sold products.

```python
pd.read_sql_query("select t2.Name, SUM(t1.Quantity) AS Quantity from sales1 AS t1 INNER JOIN products AS t2 ON t1.ProductID=t2.ProductID GROUP BY Name ORDER BY Quantity DESC LIMIT 5", con=con)
```

<br>



### `Q-2:` Find sales man who sold most no of products.



```python
pd.read_sql_query("Select FirstName, LastName, SUM(Quantity) AS Quantity From employees INNER JOIN sales1 ON employees.EmployeeID=sales1.SalesPersonID GROUP BY FirstName, LastName ORDER BY Quantity DESC LIMIT 1", con=con)
```

### `Q-3:` Sales man name who has most no of unique customer.


```python
pd.read_sql_query("Select employees.FirstName AS employee_first, employees.LastName AS employee_last, COUNT(DISTINCT sales1.CustomerID) From employees INNER JOIN sales1 ON employees.EmployeeID=sales1.SalesPersonID INNER JOIN customers ON sales1.CustomerID=customers.CustomerID GROUP BY employee_first, employee_last", con=con)
```


### `Q-4:` Sales man who has generated most revenue. Show top 5.


```python
pd.read_sql_query("SELECT employees.FirstName, employees.LastName, SUM(sales1.Quantity*products.Price) AS revenue FROM sales1 INNER JOIN products ON sales1.ProductID=products.ProductID INNER JOIN employees ON sales1.SalesPersonID=employees.EmployeeID GROUP BY FirstName, LastName ORDER BY revenue DESC LIMIT 5", con=con)
```


### `Q-5:` List all customers who have made more than 10 purchases.


```python
pd.read_sql_query("SELECT FirstName, LastName, COUNT(*) AS purchases FROM sales1 INNER JOIN customers ON sales1.SalesPersonID=customers.CustomerID GROUP BY FirstName, LastName HAVING COUNT(*)>10 ORDER BY purchases DESC", con=con)
```
