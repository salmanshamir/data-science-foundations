### Problem 6

What is the average weight difference between male and female athletes in the Olympics who have won a medal in the same event?

```python
pd.read_sql_query("SELECT * FROM athlete WHERE Height > (SELECT AVG(Height) FROM athlete WHERE Year=2008) AND Year=2008 AND Medal='Gold'", con=con)
```

```python
pd.read_sql_query("SELECT name FROM athlete WHERE Weight > (SELECT AVG(Weight) FROM athlete WHERE Year=2016 AND Medal IS NOT null) AND Year=2008 AND Medal IS NOT null AND Sport='Basketball'", con=con)
```

```python
pd.read_sql_query("SELECT name FROM athlete WHERE Medal IS NOT null AND Year IN (2008, 2016) AND Sport='Swimming'", con=con)
```

```python
pd.read_sql_query("SELECT Team FROM athlete WHERE Medal IS NOT null GROUP BY Year HAVING COUNT(*) > 50", con=con)
```

```python

pd.read_sql_query("SELECT Name FROM (SELECT Name, Year FROM athlete WHERE Medal IS NOT null GROUP BY Name, Year HAVING COUNT(DISTINCT Sport)>1) AS tab", con=con)
```

```python

a = "SELECT AVG(weight_difference) AS average_weight_difference FROM(SELECT Sport, AVG(CASE WHEN Sex = 'M' THEN Weight END) - AVG(CASE WHEN Sex = 'F' THEN Weight END) AS weight_difference FROM athlete WHERE Medal IS NOT NULL GROUP BY Sport) AS result"

pd.read_sql_query(a, con=con)
```
